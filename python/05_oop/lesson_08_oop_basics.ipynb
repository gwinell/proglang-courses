## 2. Класс — шаблон, объект — отдельная вещь

**Класс** — это описание, по которому Python создаёт объекты. Например, класс `Character` задаёт общую форму игрового персонажа.

**Объект** — один конкретный персонаж, созданный по этому описанию. У объектов одного класса есть одинаковый набор данных, но значения могут различаться.

![Класс Character как шаблон и два объекта с разными значениями](assets/class-and-objects.png)

Аналогия: класс — форма для изготовления деталей, а объекты — отдельные детали, получившиеся по этой форме. Форма общая, но каждая деталь — отдельная.

## 5. `__init__` задаёт начальные данные

Часто объекту сразу нужны имя и здоровье. Метод `__init__` **инициализирует уже созданный объект**: Python автоматически вызывает его при обычном создании экземпляра.

Сейчас запомни назначение: здесь объект получает начальные атрибуты. Метод не должен возвращать другое значение вместо `None`.

![Части класса и шаги создания объекта](assets/class-anatomy-init.png)


```python
class Character:
    def __init__(self, name, health):
        self.name = name
        self.health = health

hero1 = Character("Alex", 100)
hero2 = Character("Mia", 80)

print(hero1.name, hero1.health)
print(hero2.name, hero2.health)
```

Когда Python видит `Character("Alex", 100)`, он создаёт новый объект и запускает для него `__init__`. Значение `"Alex"` передаётся параметру `name`, а `100` — параметру `health`. Затем значения записываются в объект.

Пока можно читать `self` как «этот объект». Именно так каждый персонаж получает свои собственные данные.

## 6. Главное слово этого урока — `self`

В объявлении метода `__init__` первым параметром стоит `self`. Когда создаётся `hero1`, здесь `self` означает именно объект `hero1`. Когда создаётся `hero2`, `self` означает уже `hero2`.

Значит, `self.name = name` можно прочитать так: «сохрани полученное имя в атрибут `name` этого конкретного объекта».

![self указывает на конкретный объект при вызове его метода](assets/self-specific-object.png)

Пошагово:

1. `hero1 = Character("Alex", 100)` создаёт первый объект.
2. Во время его настройки `self` относится к `hero1`; в него записываются `Alex` и `100`.
3. `hero2 = Character("Mia", 80)` создаёт второй объект.
4. Теперь `self` относится к `hero2`; в него записываются `Mia` и `80`.

`self` не является общей коробкой для всех персонажей. Он помогает коду понять, с каким именно объектом мы сейчас работаем.


```python
class Character:
    def __init__(self, name, health):
        self.name = name       # имя этого объекта
        self.health = health   # здоровье этого объекта

hero1 = Character("Alex", 100)
hero2 = Character("Mia", 80)

print("hero1:", hero1.name, hero1.health)
print("hero2:", hero2.name, hero2.health)
```

Обрати внимание: в `__init__` записываем `self.name`, а справа читаем параметр `name`. Слева — место хранения в объекте; справа — значение, которое пришло при создании.

## 7. Атрибуты объекта

**Атрибут** — это данные, которые принадлежат объекту. В нашем примере `name` и `health` — атрибуты персонажа. Внутри класса мы обращаемся к ним как `self.name` и `self.health`, а снаружи — через точку: `hero1.name`.

Атрибут можно прочитать или изменить. Изменение одного объекта не должно менять такой же атрибут другого объекта.


```python
class Character:
    def __init__(self, name, health):
        self.name = name
        self.health = health

hero1 = Character("Alex", 100)
hero2 = Character("Mia", 80)

print(hero1.health)
hero1.health = 90
hero1.name = "Alex the Brave"

print(hero1.name, hero1.health)
print(hero2.name, hero2.health)
```

**Предскажи результат** до запуска: останется ли здоровье `hero2` равным 80?

Запись `hero1.health = 90` означает: «измени атрибут `health` у объекта, на который указывает `hero1`».

## 8. Методы: действия объектов

**Метод** — функция, написанная внутри класса. Обычно метод использует данные объекта через `self`. Например, каждый персонаж может показать свою информацию.

В определении метода первым параметром указывают `self`. При вызове `hero1.show_info()` Python сам связывает этот вызов с `hero1`, поэтому вручную передавать `self` в круглых скобках не нужно.


```python
class Character:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def show_info(self):
        print(f"Персонаж {self.name}, здоровье: {self.health}")

hero1 = Character("Alex", 100)
hero2 = Character("Mia", 80)

hero1.show_info()
hero2.show_info()
```

Внутри метода `self.name` означает имя того персонажа, у которого вызвали метод. Поэтому один и тот же код показывает разные значения.

### Метод с параметром

Метод может получать дополнительные значения. Например, персонаж получает урон. После `self` укажем параметр `damage`. `self` по-прежнему обозначает текущий объект, а `damage` — число, переданное при вызове.


```python
class Character:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def show_info(self):
        print(f"{self.name}: здоровье {self.health}")

    def take_damage(self, damage):
        self.health = self.health - damage

hero1 = Character("Alex", 100)
hero2 = Character("Mia", 80)

hero1.take_damage(15)
hero1.show_info()
hero2.show_info()
```

Вызов `hero1.take_damage(15)` можно прочитать так: «попроси объект `hero1` выполнить метод `take_damage`, передав урон 15». Внутри метода `self` — это `hero1`, а `damage` равен `15`.

## 10. Методы используют знакомые конструкции

Внутри метода можно пользоваться уже знакомыми `if`, `for`, списками и `return`. Новых правил у этих конструкций нет: просто теперь они находятся внутри класса и при необходимости используют `self`.

### Условие внутри метода

Добавим метод `is_alive()`. Он проверит атрибут здоровья и вернёт `True` или `False`.


```python
class Character:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def is_alive(self):
        if self.health > 0:
            return True
        else:
            return False

hero = Character("Alex", 10)
print(hero.is_alive())

hero.health = 0
print(hero.is_alive())
```

Здесь метод читает `self.health`, то есть здоровье текущего объекта, и возвращает результат условия.

### Метод, который перебирает список

У объекта может быть атрибут-список. Например, у персонажа есть инвентарь. Метод `show_items()` проходит по его предметам знакомым циклом `for`.


```python
class Character:
    def __init__(self, name, items):
        self.name = name
        self.items = items

    def show_items(self):
        print(f"Предметы персонажа {self.name}:")
        for item in self.items:
            print("-", item)

hero1 = Character("Alex", ["карта", "ключ"])
hero2 = Character("Mia", ["фонарь", "зелье", "карта"])

hero1.show_items()
hero2.show_items()
```

В этом примере каждому вызову передан отдельно созданный список, поэтому инвентари независимы. Если передать двум героям один список, `self.items = items` свяжет их с **общим** объектом.

### Метод может вернуть новое значение

Метод может возвращать результат через `return`. Тогда вызывающий код сам решает, сохранить ли его или напечатать.


```python
class Character:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def get_description(self):
        return f"{self.name}: здоровье {self.health}"

hero = Character("Mia", 80)
message = hero.get_description()
print(message)
print("Длина сообщения:", len(message))
```

Метод `get_description()` возвращает строку. Команда `print()` находится снаружи, поэтому мы можем решить, что делать с этой строкой.

## 11. Мини-проект: небольшая команда персонажей

Соберём знакомые идеи в короткую программу. У персонажа есть имя, здоровье и список предметов. Он умеет показать информацию, получить урон и показать предметы.

Сначала внимательно прочитай класс. Затем запусти ячейку и проследи, какие данные принадлежат каждому объекту.


```python
class Character:
    def __init__(self, name, health, items):
        self.name = name
        self.health = health
        self.items = items

    def show_info(self):
        print(f"{self.name}: здоровье {self.health}")

    def take_damage(self, damage):
        self.health = self.health - damage
        if self.health < 0:
            self.health = 0

    def show_items(self):
        print(f"Предметы {self.name}:")
        for item in self.items:
            print("-", item)

hero1 = Character("Alex", 100, ["карта", "ключ"])
hero2 = Character("Mia", 80, ["фонарь"])

hero1.take_damage(25)
hero1.show_info()
hero2.show_info()
hero1.show_items()
```

В `take_damage()` используется знакомый `if`: здоровье не становится меньше нуля. В `show_items()` используется `for`, чтобы показать каждый предмет.

Теперь список может хранить сами объекты. Цикл вызывает один и тот же метод у каждого персонажа, а `self` каждый раз относится к очередному объекту.


```python
team = [hero1, hero2]

for hero in team:
    hero.show_info()
```

Это список объектов: `team[0]` — `hero1`, а `team[1]` — `hero2`. На каждой итерации переменная `hero` указывает на очередной объект. Поэтому `hero.show_info()` показывает его собственное имя и здоровье.

## 12. Атрибут класса и атрибут экземпляра

Значение, записанное прямо в теле класса, — атрибут класса. `self.name` в `__init__` — атрибут конкретного экземпляра. Для обычных атрибутов при чтении Python сначала ищет значение у экземпляра, затем у класса.

В примере `school` общее, пока у конкретного ученика не появился свой атрибут с тем же именем. Изменение `StudentCard.school` видно тем экземплярам, которые не перекрыли его.


```python
class StudentCard:
    school = "Школа № 1"

    def __init__(self, name):
        self.name = name

anya = StudentCard("Аня")
boris = StudentCard("Борис")
anya.school = "Школа № 2"
StudentCard.school = "Лицей"
print(anya.name, anya.school)
print(boris.name, boris.school)
```

**Ловушка.** Если написать `items = []` в теле класса, это один общий список. Для личного инвентаря создавай `self.items = []` внутри `__init__`. Если список приходит аргументом и нужна независимость, используй `self.items = items.copy()` — для плоского списка это отдельный контейнер.

**Проверка понимания.** Создай двух героев с личными пустыми инвентарями. Добавь предмет первому. У второго должно остаться `[]`. Затем присвой `friend = first` и объясни, почему изменение через `friend` видно через `first`.

## Проверь себя

- Чем класс отличается от объекта?
- Для чего нужен `__init__`?
- Что означает `self` внутри метода?
- Как прочитать или изменить атрибут объекта?
- Как вызвать метод объекта?
- Чем запись `my_function()` отличается от `my_object.my_method()`?

Если можешь объяснить эти вопросы на примере `hero1` и `hero2`, значит, основные понятия урока уже понятны.
