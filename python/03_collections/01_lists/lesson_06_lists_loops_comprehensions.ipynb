# Урок 6. Списки в циклах и генераторы списков

Сегодня научимся проходить по списку с помощью `for` и `while`, а затем создавать новый список по значениям старого. Для короткой записи такого алгоритма познакомимся с **генераторами списков** — list comprehensions.

Здесь под генераторами имеются в виду только списки вида `[выражение for элемент in список]`. Функции-генераторы и `yield` мы пока не изучаем.

**Порядок урока:** список как последовательность → `for` и элементы → индекс и изменение списка → `while` → сравнение циклов → новый список через `append()` → генератор списка → условие-фильтр → практика.

Используем знакомые циклы и методы списков; `enumerate()` и новые формы записи разбираем отдельно.

## Карта урока

Список хранит элементы по порядку. Цикл помогает обработать каждый элемент: показать его, проверить условием, посчитать или добавить результат в другой список.

## 1. Вспоминаем список

Список — это последовательность значений. У каждого значения есть позиция — индекс, который начинается с нуля. Цикл позволяет пройти по элементам один за другим, не записывая отдельную команду для каждой позиции.

```python
animals = ["кот", "пёс", "ёж"]
print(animals)
print(animals[0])
print(animals[1])
print(animals[2])
```

**Мини-эксперимент.** Добавь в список четвёртое животное. Сколько команд `print()` пришлось бы написать вручную? В следующих примерах цикл выполнит повторяющиеся действия за нас.

## 2. Перебираем элементы списком `for`

Запись `for animal in animals:` означает: «возьми очередной элемент из `animals` и временно назови его `animal`». Тело цикла выполняется по одному разу для каждого элемента. После последнего элемента цикл заканчивается.

Переменная `animal` хранит **сам элемент**, а не его номер. На каждом круге цикла в ней будет новое значение.

![Цикл for проходит по элементам списка](assets/for-list.svg)

Стрелки показывают порядок: сначала в `animal` попадает `"кот"`, потом `"пёс"`, затем `"ёж"`. Команда с отступом запускается для каждого из них.

```python
animals = ["кот", "пёс", "ёж"]

for animal in animals:
    print("Я вижу:", animal)
```

**Предскажи результат.** Не запускай ячейку сразу: сколько строк появится? Затем добавь в список ещё один элемент и проверь.

```python
colors = ["синий", "зелёный", "жёлтый"]

for color in colors:
    print(color)
```

### Условие внутри цикла

Внутри цикла можно использовать знакомые `if`, `elif` и `else`. На каждом круге условие проверяется для текущего элемента.

```python
numbers = [4, 7, 2, 9, 6]

for number in numbers:
    if number > 5:
        print(number, "больше 5")
    else:
        print(number, "не больше 5")
```

**Попробуй.** Измени число `5`. До запуска предположи, какие элементы попадут в каждую ветку.

### Считаем, ищем и собираем результаты

В цикле можно завести переменную-счётчик до начала цикла, а внутри увеличивать её. Можно также искать нужные значения условием. Создавать новый список научимся чуть ниже.

```python
scores = [5, 3, 5, 4, 2, 5]
count = 0

for score in scores:
    if score == 5:
        count = count + 1

print("Пятёрок:", count)
```

**Мини-задание.** Измени список оценок. Проверь, правильно ли цикл считает количество пятёрок. Затем замени условие так, чтобы считать оценки меньше `4`.

## 3. Элемент и индекс — не одно и то же

При обычном переборе `for item in items` переменная `item` получает значение элемента. Например, строку `"пёс"`.

Иногда нужен не сам элемент, а его позиция: например, чтобы заменить значение по индексу. Тогда можно перебирать числа от `0` до `len(items) - 1` с помощью `range(len(items))`. В этом случае переменная цикла — индекс, а `items[index]` — элемент на этой позиции.

![Разница между элементом и индексом](assets/element-index.svg)

Запомни вопрос: нужно узнать **что лежит** в списке — используй элемент; нужно узнать **где лежит** значение или изменить ячейку — нужен индекс.

```python
animals = ["кот", "пёс", "ёж"]

# Перебираем сами значения
for animal in animals:
    print("Элемент:", animal)

# Перебираем номера позиций
for index in range(len(animals)):
    print("Индекс:", index, "Элемент:", animals[index])
```

**Предскажи.** Какое значение напечатает последняя итерация? Добавь к списку ещё одно животное и проверь.

### Меняем элементы по индексам

Если нужно заменить значения в исходном списке, перебирай индексы. Команда `items[index] = новое_значение` записывает значение в существующую позицию.

Не изменяй длину списка командами `append()` или `remove()` во время прохода по его индексам: позиции тогда могут сдвигаться. В этом примере мы меняем только значения.

```python
numbers = [2, 4, 6, 8]

for index in range(len(numbers)):
    numbers[index] = numbers[index] * 2

print(numbers)
```

**Попробуй.** Поменяй действие на `numbers[index] = numbers[index] + 1`. Предскажи новый список.

### Нужны сразу номер и значение? `enumerate()`

`enumerate(items)` даёт пары «индекс, элемент». В записи `for index, item in ...` два значения сразу получают два имени. Это называется распаковкой.

Для вывода номеров с единицы можно передать `start=1`. Такой номер удобен человеку, но уже не является индексом списка.


```python
names = ["Аня", "Борис", "Вера"]
for number, name in enumerate(names, start=1):
    print(number, name)
```

При удалении из списка элементы сдвигаются. Если одновременно идти по нему циклом, можно пропустить элементы. Для отбора создай новый список через `for` и `append()` или списковое включение.


## 4. Перебираем список с помощью `while`

В `while` мы сами управляем номером позиции. Сначала создаём переменную `index = 0`. Пока `index < len(items)`, индекс указывает на существующий элемент. В конце каждого круга обязательно увеличиваем `index` на единицу — иначе условие может оставаться истинным бесконечно.

![Цикл while и счётчик index](assets/while-index.svg)

Проверка `index < len(items)` не даёт обратиться к позиции за концом списка. После обработки элемента счётчик передвигается к следующей позиции.

```python
animals = ["кот", "пёс", "ёж"]
index = 0

while index < len(animals):
    print("Индекс:", index, "Элемент:", animals[index])
    index = index + 1
```

**Мини-эксперимент.** Измени список. Проследи на бумаге за значениями `index`: `0`, `1`, `2`… Что произойдёт, если убрать строку `index = index + 1`? Не запускай этот вариант надолго: цикл не сможет перейти к следующему элементу.

## 5. Одна задача: `for` или `while`?

Оба цикла могут вывести все элементы списка. В простом переборе `for` обычно удобнее: Python сам берёт следующий элемент, а мы не следим за счётчиком.

`while` полезен, когда нужно вручную управлять индексом или продолжать работу до выполнения некоторого условия. При `while` важно самим менять переменную, от которой зависит условие.

![Один список, два способа перебора](assets/for-vs-while.svg)

Результат двух программ одинаковый, но в варианте `while` мы дополнительно задаём начало, проверяем границу и увеличиваем индекс.

```python
animals = ["кот", "пёс", "ёж"]

print("Вариант for:")
for animal in animals:
    print(animal)

print("Вариант while:")
index = 0
while index < len(animals):
    print(animals[index])
    index = index + 1
```

**Вопрос.** Если нужно только вывести каждый элемент, какой цикл короче? А если нужно вручную менять индекс на два шага за круг?

## 6. Создаём новый список через `for` и `append()`

Частая задача — не менять старые значения, а создать **новый список** на их основе. Сначала создаём пустой список `doubled`. Затем цикл берёт значение из `numbers`, вычисляет новое значение и добавляет его в конец нового списка через `append()`.

Важно: `numbers` остаётся прежним, а результаты по одному появляются в `doubled`. Это уже знакомый цикл и знакомый метод `append()` — сейчас мы соединяем их.

![Создание нового списка с помощью for и append](assets/for-append.svg)

Для каждого элемента старого списка выполняется одно действие: вычислить `number * 2`, затем добавить ответ в новый список.

```python
numbers = [2, 4, 6]
doubled = []

for number in numbers:
    doubled.append(number * 2)

print("Старый список:", numbers)
print("Новый список:", doubled)
```

**Проследи по шагам.** После первого круга `doubled` станет `[4]`, после второго — `[4, 8]`, после третьего — `[4, 8, 12]`. Замени умножение на `number + 10` и проверь.

## 7. Генератор списка — короткая запись знакомого алгоритма

Генератор списка (list comprehension) создаёт новый список по элементам исходного. Это не новый вид цикла: внутри записи по-прежнему есть действие для каждого элемента. Просто Python записывает такой алгоритм компактнее.

Разберём запись по частям:

- квадратные скобки `[...]` означают, что результатом будет новый список;
- `number * 2` — выражение: какое значение положить в новый список;
- `for number in numbers` — откуда брать элементы и как назвать текущий элемент.

Вся запись читается так: «создай список результатов `number * 2` для каждого `number` из `numbers`».

![Тот же алгоритм короткой записью](assets/list-comprehension.svg)

Слева — уже знакомый `for` с `append()`. Справа — короткая запись той же идеи.

```python
numbers = [2, 4, 6]
doubled = [number * 2 for number in numbers]
print(doubled)
```

### Сравним две записи

Эти фрагменты делают одно и то же. Сначала исполняется обычный цикл: создаём пустой список и добавляем результат на каждом круге. В короткой записи квадратные скобки создают список, выражение задаёт новое значение, а часть `for ... in ...` перебирает исходные элементы.

```python
# Длинная запись
numbers = [2, 4, 6]
doubled = []
for number in numbers:
    doubled.append(number * 2)
print(doubled)

# Короткая запись
numbers = [2, 4, 6]
doubled = [number * 2 for number in numbers]
print(doubled)
```

## 8. Добавляем условие-фильтр `if`

Иногда в новый список нужно взять не все элементы, а только подходящие. После части `for ... in ...` можно поставить знакомое условие `if`.

Общая форма, которую сегодня используем: `[выражение for элемент in список if условие]`.

Порядок чтения такой: взять элемент → проверить условие → если условие истинно, вычислить выражение и добавить результат в новый список. Если условие ложно, этот элемент пропускается.

![Генератор списка с условием if](assets/comprehension-if.svg)

Здесь выражение оставляет число без изменения, а фильтр `if number > 0` пропускает в результат только положительные числа.

```python
numbers = [-3, 0, 2, -1, 5, 4]
positive = [number for number in numbers if number > 0]
print(positive)
```

**Мини-эксперимент.** Поменяй условие на `number >= 0`. Как изменится результат? Затем попробуй `number % 2 == 0`, чтобы оставить чётные числа.

### Сначала длинная версия, затем короткая

Фильтр тоже можно записать обычным циклом. На каждом круге `if` решает, добавлять ли текущий элемент. Короткая запись передаёт ту же последовательность действий.

```python
numbers = [-3, 0, 2, -1, 5, 4]
positive = []
for number in numbers:
    if number > 0:
        positive.append(number)
print(positive)

positive = [number for number in numbers if number > 0]
print(positive)
```

### Фильтр и выбор значения — разные задачи

`[number for number in numbers if number > 0]` **отбирает** положительные элементы. Число элементов может уменьшиться.

`[number if number > 0 else 0 for number in numbers]` создаёт результат для **каждого** элемента: положительные оставляет, остальные заменяет нулём. Здесь `if ... else ...` — выражение выбора значения, оно стоит перед `for`.

Сначала напиши обычный цикл. Сокращай запись, только если её по-прежнему легко прочитать.


```python
numbers = [-2, 0, 3, 5]
positive = [number for number in numbers if number > 0]
replaced = [number if number > 0 else 0 for number in numbers]
print(positive)
print(replaced)
```

## 9. Небольшие программы для практики

В этих примерах мы объединяем списки с уже знакомыми циклами и условиями. Сначала прочитай код и предскажи результат, затем запусти.

```python
# Считаем, сколько чисел больше 10
numbers = [4, 12, 7, 18, 10, 25]
count = 0
for number in numbers:
    if number > 10:
        count = count + 1
print("Чисел больше 10:", count)
```

Создадим новый список цен со скидкой 10%. Умножаем каждую цену на `0.9`.

```python
prices = [100, 250, 80, 400]
new_prices = [price * 0.9 for price in prices]
print(new_prices)
```

Отфильтруем оценки, которые не ниже четырёх.

```python
grades = [5, 3, 4, 2, 5, 3, 4]
good_grades = [grade for grade in grades if grade >= 4]
print(good_grades)
```

### Дополнительно: таблица как список строк

В `board[row][column]` сначала выбирается строка таблицы, затем элемент в этой строке. Индексы начинаются с нуля.

`[[0] * 3] * 2` повторяет ссылку на **одну** строку. Чтобы строки менялись независимо, создавай каждую заново: `[[0] * 3 for row in range(2)]`.


```python
shared_rows = [[0] * 3] * 2
shared_rows[0][1] = 7
print(shared_rows)

board = [[0] * 3 for row in range(2)]
board[0][1] = 7
print(board)
```

## Короткое повторение

- `for item in items` перебирает сами значения по порядку.
- `range(len(items))` помогает пройти по индексам; `items[index]` — элемент на этой позиции.
- Чтобы заменить значения в списке, удобно использовать индексы.
- В `while` со счётчиком нужно проверить `index < len(items)` и менять индекс на каждом круге.
- Для простого перебора списка чаще удобнее `for`; `while` даёт ручной контроль над условием и индексом.
- Новый список можно собрать циклом: создать `result = []`, вычислить результат и вызвать `result.append(...)`.
- `[выражение for элемент in список]` — короткая запись такого алгоритма.
- `[выражение for элемент in список if условие]` добавляет фильтр: в новый список проходят только подходящие элементы.

**Важно:** генераторы списков в этом уроке создают списки. Generator functions, `yield` и generator expressions здесь не рассматриваются.

**Сюжет.** Для широких этикеток подходят названия длиннее четырёх символов.
