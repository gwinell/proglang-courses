## 1. Кортеж: значения в фиксированном порядке

Кортеж похож на список: в нём можно хранить несколько значений, обращаться к ним по индексу и узнавать их количество с помощью `len()`.

Кортеж записывают в круглых скобках. После создания его элементы нельзя заменить или удалить.


```python
point = (4, 7)
print(point[0])
print(len(point))
```

Сравним со списком. В списке элемент можно изменить, а в кортеже — нельзя.


```python
colors_list = ["синий", "зелёный"]
colors_list[0] = "красный"
print(colors_list)

colors_tuple = ("синий", "зелёный")
# Попробуй раскомментировать строку ниже:
# colors_tuple[0] = "красный"
```

### Кортеж из одного элемента

У кортежа из одного элемента обязательно должна быть запятая. Без неё круглые скобки просто группируют значение.


```python
one_item_tuple = ("Python",)
not_a_tuple = ("Python")

print(type(one_item_tuple))
print(type(not_a_tuple))
```

### Распаковываем кортеж

Если переменных столько же, сколько значений в кортеже, можно присвоить значения нескольким переменным сразу.


```python
point = (4, 7)
x, y = point

print(f"x = {x}")
print(f"y = {y}")
```

Это удобно, когда функция или выражение возвращает несколько связанных значений. Пока запомни сам приём распаковки — к нему можно будет вернуться в следующих проектах.

## 2. Словарь: значение по ключу

Список ищет элемент по номеру. Словарь ищет значение по **ключу**: например, по имени поля `"name"` можно получить значение `"Аня"`.

Словарь записывают в фигурных скобках. Пары ключ-значение разделяются запятыми, а между ключом и значением ставится двоеточие.


```python
student = {
    "name": "Аня",
    "grade": 5,
}

print(student["name"])
print(student["grade"])
```

![Ключи и значения в словаре](assets/dictionary.png)

Ключи помогают описать, что означает каждое значение. Для нового поля можно записать новое значение по новому ключу.


```python
student = {"name": "Аня", "grade": 5}

student["city"] = "Казань"       # добавили новую пару
student["grade"] = 6             # обновили значение по ключу

print(student)
```

У каждого ключа в словаре одно значение. Если записать значение по уже существующему ключу, старое значение заменится.

Чтобы безопасно получить значение, даже если ключа может не быть, используй `get()`. Если ключ не найден, `get()` вернёт `None` или запасное значение, которое ты укажешь.


```python
student = {"name": "Аня", "grade": 5}

print(student.get("name"))
print(student.get("city"))
print(student.get("city", "город не указан"))
```

Удалить пару можно командой `del` с ключом.


```python
student = {"name": "Аня", "grade": 5, "city": "Казань"}
del student["city"]
print(student)
```

## 3. Перебираем словарь

Цикл `for` по словарю перебирает его ключи. По каждому ключу можно получить значение.


```python
scores = {"Аня": 5, "Борис": 4, "Вера": 5}

for name in scores:
    print(name, scores[name])
```

Метод `items()` даёт пары «ключ и значение». Их можно сразу распаковать в цикле.


```python
scores = {"Аня": 5, "Борис": 4, "Вера": 5}

for name, score in scores.items():
    print(f"{name}: оценка {score}")
```

Если нужны только ключи или только значения, есть методы `keys()` и `values()`.


```python
scores = {"Аня": 5, "Борис": 4, "Вера": 5}

print(list(scores.keys()))
print(list(scores.values()))
```

## 4. Множество: только уникальные значения

Множество хранит каждый элемент только один раз. Оно удобно, когда нужно убрать повторы или быстро проверить, встречается ли значение.

Множество создают с помощью `set()` или фигурных скобок с элементами. Пустое множество создаётся как `set()`: пустые `{}` означают пустой словарь.


```python
numbers = {2, 4, 2, 6, 4}
print(numbers)
print(len(numbers))

empty_set = set()
empty_dictionary = {}
print(type(empty_set))
print(type(empty_dictionary))
```

Порядок элементов множества не задан. Не обращайся к ним по индексам, как в списке или кортеже.

![Множество оставляет только уникальные значения](assets/set_unique.png)

Создать множество можно из строки или списка. Повторяющиеся элементы останутся только по одному.


```python
letters = set("малина")
print(letters)

colors = ["синий", "зелёный", "синий", "жёлтый"]
unique_colors = set(colors)
print(unique_colors)
```

### Добавляем и удаляем элементы

`add()` добавляет элемент, а `discard()` удаляет его, если он есть. Если такого элемента нет, `discard()` ничего не меняет.


```python
animals = {"кот", "пёс"}
animals.add("ёж")
animals.discard("пёс")

print(animals)
```

Оператор `in` проверяет, входит ли элемент в множество.


```python
members = {"Аня", "Борис", "Вера"}
print("Вера" in members)
print("Глеб" in members)
```

## 5. Сравниваем множества

Пусть есть два кружка по интересам. Множества помогут найти тех, кто записался в оба кружка, только в один или хотя бы в один.


```python
chess = {"Аня", "Борис", "Вера"}
robotics = {"Вера", "Глеб", "Аня"}

both = chess & robotics       # пересечение: есть в обоих множествах
any_club = chess | robotics   # объединение: есть хотя бы в одном
only_chess = chess - robotics # есть в chess, но нет в robotics

print("Оба кружка:", both)
print("Хотя бы один:", any_club)
print("Только шахматы:", only_chess)
```

Запомни обозначения:

- `a | b` — объединение;
- `a & b` — пересечение;
- `a - b` — элементы, которые есть в `a`, но отсутствуют в `b`.

## 6. Как выбрать коллекцию

| Коллекция | Порядок важен? | Можно менять после создания? | Повторы? | Как получить элемент |
|---|---:|---:|---:|---|
| Список `list` | Да | Да | Да | По индексу |
| Кортеж `tuple` | Да | Нет | Да | По индексу |
| Словарь `dict` | Хранит пары ключ-значение | Да | Ключи уникальны | По ключу |
| Множество `set` | Нет | Да | Нет | Проверить принадлежность |
