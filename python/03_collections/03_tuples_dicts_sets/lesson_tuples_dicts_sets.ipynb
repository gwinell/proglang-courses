## 1. Кортеж: значения в фиксированном порядке

Кортеж — это несколько значений, собранных вместе в определённом порядке. Он похож на список: у каждого значения есть индекс, а len() считает элементы. Главное отличие: отдельные элементы кортежа после создания заменить или удалить нельзя.

Например, точка на карте — это пара координат x и y. Сначала записывают x, потом y: если поменять их местами, получится другая точка.

![Кортеж point = (4, 7): у элементов индексы 0 и 1](assets/tuple_indexes.svg)

Первое значение имеет индекс 0, второе — индекс 1. Поэтому point[0] возвращает 4.

```python
point = (4, 7)
print(point[0])
print(len(point))
```

Сравним со списком. В списке элемент можно изменить, а в кортеже — нельзя.

В кортеже такая замена запрещена. Попробуй предсказать, что произойдёт, если убрать знак комментария перед последней строкой и запустить ячейку. Python покажет ошибку: кортеж неизменяемый.

Это значит, что нельзя заменить или удалить элемент внутри уже созданного кортежа. Но переменной можно присвоить новый кортеж: point = (5, 7). Так переменная будет хранить другое значение.


```python
colors_list = ["синий", "зелёный"]
colors_list[0] = "красный"
print(colors_list)

colors_tuple = ("синий", "зелёный")
# Попробуй раскомментировать строку ниже:
# colors_tuple[0] = "красный"
```

### Кортеж из одного элемента

Для кортежа из одного элемента нужна запятая. Скобки сами по себе могут просто группировать значение: ("Python",) — кортеж, а ("Python") — строка. Запусти пример и сравни типы данных.

```python
one_item_tuple = ("Python",)
not_a_tuple = ("Python")

print(type(one_item_tuple))
print(type(not_a_tuple))
```

### Распаковываем кортеж

Распаковка раскладывает значения по переменным: первое значение попадёт в x, второе — в y. Количество переменных должно совпадать с количеством элементов кортежа, иначе Python сообщит об ошибке.

До запуска попробуй предсказать значения x и y.

```python
point = (4, 7)
x, y = point

print(f"x = {x}")
print(f"y = {y}")
```

### Неизменяемый кортеж может ссылаться на изменяемый список

В кортеже нельзя заменить его элемент, но объект внутри может меняться по своим правилам. В примере элемент `record[1]` всё время ссылается на один список, а в этот список добавляется новая оценка.


```python
record = ("Аня", [5, 4])
record[1].append(5)
print(record)
```

Это удобно, когда функция возвращает несколько связанных значений, например координаты x и y.

- len(point) считает число элементов.
- Элемент кортежа нельзя заменить по индексу.
- Для одного элемента нужна запятая: ("кот",).
- При распаковке значения распределяются слева направо.

## 2. Словарь: значение по ключу

Список ищет элемент по номеру. Словарь ищет значение по **ключу**: например, по имени поля "name" можно получить значение "Аня". Представь анкету: ключ — подпись поля, значение — ответ.

Словарь записывают в фигурных скобках. Пары «ключ: значение» разделяются запятыми, между ключом и значением ставится двоеточие.

```python
student = {
    "name": "Аня",
    "grade": 5,
}

print(student["name"])
print(student["grade"])
```

![Ключи и значения в словаре](assets/dictionary.png)

Каждый ключ объясняет значение: "name" — подпись для имени, "grade" — подпись для оценки. По ключу находят нужное значение. До запуска попробуй предсказать результат.

```python
student = {"name": "Аня", "grade": 5}

student["city"] = "Казань"       # добавили новую пару
student["grade"] = 6             # обновили значение по ключу

print(student)
```

У каждого ключа словаря одно текущее значение. Запись по уже существующему ключу заменяет его, а запись по новому ключу добавляет пару. Значения могут повторяться — у нескольких учеников может быть одинаковая оценка.

Если ключа может не быть, используй get(). Обычное обращение к отсутствующему ключу вызовет ошибку; get() вернёт None или указанное запасное значение. Предскажи результаты трёх вызовов get() до запуска.

```python
student = {"name": "Аня", "grade": 5}

print(student.get("name"))
print(student.get("city"))
print(student.get("city", "город не указан"))
```

Удалить пару можно командой del с ключом. В примере удаляется только пара с ключом city; остальные поля остаются. Если ключа нет, del выдаст ошибку. get() безопасно читает отсутствующий ключ, а del удаляет существующую пару.

```python
student = {"name": "Аня", "grade": 5, "city": "Казань"}
del student["city"]
print(student)
```

### `in` в словаре проверяет ключи

`"name" in student` проверяет ключ. Для поиска среди значений используй `student.values()`. `get()` читает отсутствующий ключ с запасным результатом, но не добавляет его.

Ключ должен быть **хешируемым**: Python должен уметь получить для него устойчивое хеш-значение. Строки и целые числа подходят, список — нет. Кортеж подходит, только если все его элементы хешируемы; кортеж со списком не подойдёт. Для первых задач используй строковые ключи.


```python
student = {"name": "Аня", "score": 5}
print("name" in student)
print("Аня" in student)
print("Аня" in student.values())
print(student.get("city", "не указан"))
print("city" in student)
```

## 3. Перебираем словарь

Цикл `for` по словарю перебирает его ключи. По каждому ключу можно получить значение.

Цикл for по словарю проходит по ключам. На каждом шаге name получает имя, а scores[name] находит оценку. Попробуй угадать число строк вывода до запуска.


```python
scores = {"Аня": 5, "Борис": 4, "Вера": 5}

for name in scores:
    print(name, scores[name])
```

Метод items() даёт пары «ключ и значение». Их можно сразу распаковать в две переменные: name получает ключ, score — оценку. В предыдущем примере оценку находили отдельно через scores[name].

```python
scores = {"Аня": 5, "Борис": 4, "Вера": 5}

for name, score in scores.items():
    print(f"{name}: оценка {score}")
```

Если нужны только ключи, используй keys(); только значения — values(). list(...) показывает результат как обычный список. Предскажи, где будут имена, а где оценки.

### Повторение про словарь

- Ключ — подпись, значение — данные.
- Новый ключ добавляет пару, существующий ключ обновляет значение.
- get() помогает безопасно получить ответ, если ключа нет.
- Цикл по словарю перебирает ключи; items() даёт пару.

```python
scores = {"Аня": 5, "Борис": 4, "Вера": 5}

print(list(scores.keys()))
print(list(scores.values()))
```

## 4. Множество: только уникальные значения

Множество хранит каждый элемент только один раз. Оно удобно, когда нужно убрать повторы или проверить наличие значения. Представь список гостей, где повторное имя оставляют только один раз.

Множество создают с помощью set() или фигурных скобок с элементами. Пустое множество — set(); пустые фигурные скобки {} создают пустой словарь. До запуска посчитай, сколько разных чисел останется в примере.

```python
numbers = {2, 4, 2, 6, 4}
print(numbers)
print(len(numbers))

empty_set = set()
empty_dictionary = {}
print(type(empty_set))
print(type(empty_dictionary))
```

Порядок множества не задан: элементы могут выводиться в разной последовательности, это нормально. У множества нет индексов.

Множество можно создать из списка или строки; повторы исчезнут. В слове "малина" каждая буква станет элементом, а повторная буква "а" останется один раз.

![Множество оставляет только уникальные значения](assets/set_unique.png)

```python
letters = set("малина")
print(letters)

colors = ["синий", "зелёный", "синий", "жёлтый"]
unique_colors = set(colors)
print(unique_colors)
```

### Добавляем и удаляем элементы

add() добавляет элемент. Повторно добавить копию не получится, потому что множество хранит только уникальные значения. discard() удаляет элемент, если он есть; если его нет, множество не меняется.

Представь набор карточек: add кладёт карточку, discard убирает её. Какое животное останется после команд?

```python
animals = {"кот", "пёс"}
animals.add("ёж")
animals.discard("пёс")

print(animals)
```

Оператор in проверяет, входит ли значение в множество. Ответ — True или False. Прочитай проверку как вопрос: «Вера входит в группу?» Предскажи два ответа до запуска.

```python
members = {"Аня", "Борис", "Вера"}
print("Вера" in members)
print("Глеб" in members)
```

## 5. Сравниваем множества

Представь два кружка по интересам: шахматы и робототехника. Множества помогут найти участников обоих кружков, хотя бы одного или только одного.

В примере Аня и Вера есть в обоих множествах, Борис — только в шахматах, Глеб — только в робототехнике. Попробуй найти эти группы до запуска.

```python
chess = {"Аня", "Борис", "Вера"}
robotics = {"Вера", "Глеб", "Аня"}

both = chess & robotics       # пересечение: есть в обоих множествах
any_club = chess | robotics   # объединение: есть хотя бы в одном
only_chess = chess - robotics # есть в chess, но нет в robotics

print("Оба кружка:", both)
print("Хотя бы один:", any_club)
print("Только шахматы:", only_chess)
```

Запомни три действия:

- a | b — объединение: есть хотя бы в одном множестве.
- a & b — пересечение: есть в обоих множествах.
- a - b — разность: есть в первом множестве, но нет во втором.

Порядок важен для разности: chess - robotics — только шахматы; robotics - chess — только робототехника. Вывод может показывать элементы в разной последовательности, но состав ответа тот же.

### Одна задача для словаря: частоты слов

Ключ — слово, значение — сколько раз оно встретилось. Для ещё не встречавшегося слова `get(word, 0)` даёт 0. Затем счётчик увеличивается и сохраняется обратно.


```python
words = ["кот", "пёс", "кот", "кот"]
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
print(counts)
```

**Самостоятельно.** Посчитай слова в строке `"дом лес дом"` через `split()`. Ожидается `{'дом': 2, 'лес': 1}`. Для пустой строки — `{}`. Если нужны только разные слова, выбери `set`; если число повторений — `dict`.

У множества, как и у ключей словаря, элементы должны быть хешируемыми. У словаря сохраняется порядок добавления ключей, но это не сортировка; у множества заданного порядка нет.

**Ограничения.** Имя/вид непустые до 40 символов; 0 ≤ возраст ≤ 30.

**Примеры**

| Входные данные | Выходные данные |
|---|---|
| <pre>Барсик<br>кот<br>2</pre> | <pre>Имя: Барсик<br>Вид: кот<br>Возраст: 2</pre> |
| <pre>Луч<br>пёс<br>0</pre> | <pre>Имя: Луч<br>Вид: пёс<br>Возраст: 0</pre> |

### ДЗ-11.4. Список гостей · повторение

**Условие.** Уберите пробелы по краям имён, исключите повторы и выведите число разных гостей. Регистр важен.

### ДЗ-11.5. Покупки команды · по выбору

**Сюжет.** Записи о товаре могут повторяться.

**Условие.** Сложите количества одинаковых товаров в словаре. Выведите товары в порядке первого появления.

**Формат данных**

| Входные данные | Выходные данные |
|---|---|
| n, затем для каждой записи две строки: название, количество. | По записи «название количество» на строке; при n=0 вывод пустой. |

**Ограничения.** 0 ≤ n ≤ 50; название непустое до 30 символов; 0 ≤ количество ≤ 100.

**Примеры**

| Входные данные | Выходные данные |
|---|---|
| <pre>3<br>чай<br>2<br>сок<br>1<br>чай<br>3</pre> | <pre>чай 5<br>сок 1</pre> |
| <pre>0</pre> | <em>Нет вывода</em> |

**Пояснение.** Повторный товар увеличивает количество, но не создаёт новую строку. Для порядка можно хранить отдельный список новых названий.

**Источник.** Авторская задача курса.
