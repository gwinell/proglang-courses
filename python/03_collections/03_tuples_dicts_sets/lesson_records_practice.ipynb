## 1. Одна запись и много записей

Словарь удобно описывает один объект: у книги есть название, автор и год. Если книг несколько, список сохраняет их порядок, а каждый его элемент остаётся отдельной записью. Индексы списка начинаются с нуля.

В словаре записи обращаемся к полю по ключу. В списке записей сначала выбираем элемент, затем его поле: `books[1]["title"]`.

```python
books = [
    {"title": "Снежная королева", "author": "Андерсен", "year": 1844},
    {"title": "Муму", "author": "Тургенев", "year": 1852},
]

print(books[0]["title"])
print(books[1]["year"])
```

**Проверь себя.** Сколько здесь словарей? Что вернёт `books[0]`? Почему `books[1]["year"]` — это число, а не словарь?

Обычное присваивание `book = books[0]` не создаёт копию записи: оба имени обращаются к одному словарю. Если изменить `book["year"]`, изменится и значение внутри `books`. Это ожидаемо, когда мы хотим обновить найденную запись.

```python
book = books[0]
book["year"] = 1845
print(books[0])
```

## 2. Данные внутри данных

В записи поле может содержать список, например жанры или участников. Внешний словарь отвечает за поля книги, а список `genres` хранит несколько значений одного поля. Можно встретить и словарь внутри записи — например, координаты или адрес. Читай выражение снаружи внутрь: сначала выбери запись, потом поле, затем элемент или вложенный ключ.

Не вкладывай коллекции без необходимости: каждый уровень должен отвечать на отдельный вопрос о данных.

```python
library = {
    "books": [
        {"title": "Снежная королева", "genres": ["сказка", "приключения"]},
        {"title": "Муму", "genres": ["рассказ"]},
    ],
    "address": {"city": "Тула", "street": "Садовая"},
}

print(library["books"][0]["genres"][1])
print(library["address"]["city"])
```

## 3. Поиск и изменение записи

Чтобы найти запись по `id`, перебирай список и сравнивай поле каждой записи с искомым значением. Идентификатор в этом примере уникален. Если совпадения нет, переменная `found` остаётся равна `None`; не обращайся к полям до проверки.

Сначала покажем поиск. После успешного поиска можно обновить поле найденного словаря; отдельное повторное присваивание в списке не нужно, потому что `found` ссылается на эту запись.

```python
students = [
    {"id": 101, "name": "Лена", "score": 78},
    {"id": 102, "name": "Илья", "score": 91},
    {"id": 103, "name": "Маша", "score": 84},
]

wanted_id = 102
found = None
for student in students:
    if student["id"] == wanted_id:
        found = student
        break

if found is None:
    print("Ученик не найден")
else:
    found["score"] = 95
    print(found)
```

## 4. Фильтрация и сводные значения

Фильтрация создаёт новый список только из записей, прошедших условие. Начинай с пустого списка, добавляй подходящие записи и отдельно считай итог. Так алгоритм остаётся видимым: один проход, одна проверка, одно действие.

Если список пуст, сумма оценок равна нулю, а среднего значения нет: делить на количество записей нельзя. Условие должно явно определить, что показать в этом случае.

```python
passed = []
total = 0
for student in students:
    if student["score"] >= 80:
        passed.append(student)
        total += student["score"]

print("Сдали:", len(passed))
print("Сумма баллов:", total)
if passed:
    print("Средний балл:", total / len(passed))
else:
    print("Нет учеников для расчёта среднего")
```

- Один словарь описывает одну запись; список объединяет много записей.

- Перед расчётом среднего проверь, что записей больше нуля.
