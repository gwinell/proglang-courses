# Дополнительно: итераторы и выражения-генераторы

Нужно знать списки, `for`, списковые включения и функции. Здесь научимся различать готовую коллекцию и однократный поток значений. Этот блок можно отложить до закрепления основных уроков.

Примеры опираются на идеи § 5.4 и 6.5 части I книги С. Балакирева (2026); формулировки и задания подготовлены для этого курса.


## 1. Что цикл получает от последовательности

**Итерируемый объект** умеет предоставлять элементы для перебора: список, строка, `range`. `iter(data)` получает **итератор**, который помнит продвижение по элементам. `next(iterator)` получает следующий элемент.

После конца `next()` поднимает `StopIteration`. У `next(iterator, default)` можно задать запасной результат — это удобно для первого эксперимента.


```python
names = ["Аня", "Борис"]
iterator = iter(names)
print(next(iterator))
print(next(iterator))
print(next(iterator, "Конец"))
print(next(iterator, "Конец"))
```

Список можно перебирать снова: новый `for` получает новый итератор. У уже исчерпанного итератора новых элементов нет. `range` тоже можно перебирать повторно: он не является одноразовым генератором.


```python
numbers = [2, 4]
iterator = iter(numbers)
print(list(iterator))
print(list(iterator))
print(list(numbers))
```

## 2. Квадратные и круглые скобки

`[number * 2 for number in range(3)]` сразу создаёт список. `(number * 2 for number in range(3))` создаёт генератор: результаты вычисляются по мере запроса.

| Свойство | Списковое включение | Выражение-генератор |
|---|---|---|
| Результат | готовый список | генератор, то есть итератор |
| Повторный обход | да | продолжает с текущего места |
| `len()` и индекс | есть | обычно нет |
| Когда вычисляются элементы | при создании списка | при получении очередного элемента |

После исчерпания генератора повторный обход пуст. Круглые скобки здесь не создают кортеж.


```python
ready = [number * 2 for number in range(3)]
stream = (number * 2 for number in range(3))
print(ready)
print(next(stream))
print(list(stream))
print(list(stream))
```

## 3. `zip`, `any` и `all` в понятной задаче

`zip(first, second)` даёт пары элементов и по умолчанию заканчивается на более коротком источнике. Его результат — итератор. Для таблицы, где длины обязаны совпадать, в Python 3.10+ есть `zip(..., strict=True)`.

`any(conditions)` отвечает, есть ли хотя бы одно истинное значение; `all(conditions)` — все ли истинны. Для пустого источника `any` даёт `False`, `all` даёт `True`. Если правило требует хотя бы одного элемента, проверяй непустоту отдельно.


```python
names = ["Аня", "Борис"]
scores = [5, 4]
for name, score in zip(names, scores):
    print(name, score)
print(any(score == 5 for score in scores))
print(all(score >= 3 for score in scores))
print(any([]), all([]))
```
