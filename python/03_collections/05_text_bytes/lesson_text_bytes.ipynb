# Текст и байты: str, bytes и UTF-8

До этого урока мы работали с текстом Python типа str. Теперь разберём, как текст превращается в байты при сохранении и передаче, почему кириллица занимает разное число байтов и как явно преобразовать данные.

## Результаты

После занятия ты сможешь отличить текст от байтов, кодировать строку в UTF-8, декодировать корректные байты и объяснить ошибку при неверной последовательности.

## 1. Символы и байты — разные данные

str — последовательность символов, с которой обычно работает человек. bytes — последовательность чисел от 0 до 255. Буква A в UTF-8 занимает один байт, а многие кириллические буквы — два. Поэтому длина в символах и длина в байтах могут различаться.

Байтовый литерал записывается с префиксом b; в нём можно напрямую указывать только ASCII-символы.

```python
text = "Aя"
data = b"A"

print(type(text).__name__)
print(type(data).__name__)
print(len(text))
print(len(data))
```

**Предскажи:** длина текста Aя равна двум символам. После выполнения кодирования в следующем примере увидим три байта: один для A и два для я. Не используй len() для оценки длины текста в байтах, пока явно не получил объект bytes.

## 2. Кодирование строки в UTF-8

Кодирование — преобразование str в байты. UTF-8 — распространённая кодировка для хранения текста. Вызов encode возвращает новый объект bytes; исходная строка не меняется.

```python
message = "Привет, Python!"
encoded = message.encode("utf-8")

print(type(encoded).__name__)
print(encoded)
print(len(message), "символов")
print(len(encoded), "байт")
```

## 3. Декодирование байтов

Декодирование — обратное преобразование: нужно указать ту же кодировку, которой байты были созданы. decode возвращает str. Когда читаешь текстовый файл через open(..., encoding="utf-8"), Python делает это преобразование за тебя.

```python
restored = encoded.decode("utf-8")
print(restored)
print(restored == message)
```

## 4. Неверные байты и явные правила ошибки

Не каждая последовательность байтов является допустимым UTF-8. По умолчанию decode поднимет UnicodeDecodeError. Не подавляй её вслепую: сначала выясни, какая кодировка и какие данные ожидаются. Если допустима потеря неизвестных символов, параметр errors="replace" явно заменит их на �; это меняет данные и годится только если такое поведение разрешено.

```python
invalid = bytes([0xFF])

try:
    print(invalid.decode("utf-8"))
except UnicodeDecodeError as error:
    print(type(error).__name__)

print(invalid.decode("utf-8", errors="replace"))
```

## 5. Где это встречается

- Строки программы и пользовательский текст: str.
- Байтовые файлы и сетевые данные: часто bytes.
- Текстовый файл с UTF-8: Python кодирует текст при записи и декодирует байты при чтении благодаря encoding="utf-8".
- Если получил bytes, сначала выясни формат и кодировку, затем вызывай decode.
