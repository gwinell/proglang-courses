## 1. Метод строки — действие над текстом

У строки есть **методы** — готовые действия, которые вызываются после точки. Например, `word.upper()` создаёт строку с заглавными буквами.

Обрати внимание на скобки после имени метода: они означают, что мы его вызываем.


```python
word = "Python"
print(word.upper())
print(word.lower())
```

Методы не меняют исходную строку. Они возвращают новое значение. Сохрани его в переменную, если хочешь использовать результат дальше.


```python
name = "аня"
uppercase_name = name.upper()

print(uppercase_name)
print(name)
```

## 2. Регистр и пробелы по краям

- `upper()` делает буквы заглавными.
- `lower()` делает буквы строчными.
- `title()` делает первую букву каждого слова заглавной.
- `strip()` убирает пробелы и другие пробельные символы **с начала и конца** строки.

`strip()` не убирает пробел между словами.


```python
title = "путешествие по питону"
print(title.title())

message = "   Привет, мир!   "
print(message.strip())
```

![Метод strip убирает пробелы только по краям строки](assets/strip.png)

Ввод пользователя часто содержит случайные пробелы. `strip()` помогает привести такой ввод в порядок:


```python
answer = "  да  "
answer = answer.strip()
print(f"Ответ: {answer}!")
```

Есть и два более точных варианта: `lstrip()` убирает пробелы слева, а `rstrip()` — справа.


```python
text = "   Python   "
print(text.lstrip())
print(text.rstrip())
```

## 3. Проверяем и ищем текст

Оператор `in` проверяет, встречается ли подстрока. Результат — `True` или `False`.


```python
sentence = "Я изучаю Python"
print("Python" in sentence)
print("Java" in sentence)
```

Метод `find()` возвращает индекс первого найденного фрагмента. Если фрагмент не найден, результатом будет `-1`.


```python
sentence = "банан и ананас"
print(sentence.find("ан"))
print(sentence.find("арбуз"))
```

### Почему нельзя проверять поиск как `if text.find(...)`

Результат `find()` — индекс: 0 означает «нашли в начале», а -1 — «не нашли». В условии 0 считается ложным, а -1 — истинным. Поэтому для наличия фрагмента используй `part in text`, а для позиции — `position != -1`.


```python
text = "Python и игры"
position = text.find("Python")
print(position)
print("Python" in text)
print(text.find("Java") == -1)
```

Метод `count()` считает, сколько раз встречается фрагмент. А `startswith()` и `endswith()` проверяют начало и конец строки.


```python
filename = "report.txt"
print(filename.count("t"))
print(filename.startswith("report"))
print(filename.endswith(".txt"))
```

### Проверим ввод

Представим, что в коде есть команда. Проверим, содержит ли ответ пользователя нужное слово:


```python
message = input("Напиши сообщение: ")
if "привет" in message.lower():
    print("Ты поздоровался!")
else:
    print("Слова «привет» в сообщении нет.")
```

## 4. Заменяем часть строки

`replace(что_найти, на_что_заменить)` возвращает новую строку, в которой найденные фрагменты заменены.


```python
sentence = "Я люблю кошек. Кошки очень умные!"
new_sentence = sentence.replace("кошек", "собак")
print(new_sentence)
print(sentence)
```

В исходной строке регистр важен: `replace("кошек", ...)` заменяет только точное совпадение `кошек`, но не `Кошки`. Метод также не меняет исходную строку.

## 5. Разбиваем строку на части: `split()`

`split()` превращает строку в список частей. Если вызвать его без аргумента, Python разделит текст по пробелам и другим пробельным символам.


```python
sentence = "Я изучаю строки в Python"
words = sentence.split()
print(words)
print(len(words))
```

Можно указать разделитель явно. Например, если значения разделены запятой и пробелом, передай `", "`.


```python
drinks_text = "чай, сок, вода"
drinks = drinks_text.split(", ")
print(drinks)
print(drinks[0])
```

### `split()` и `split(" ")` работают по-разному

Без аргумента `split()` объединяет подряд идущие пробельные символы и убирает пустые края. С явным разделителем каждое совпадение делит строку, поэтому могут появляться пустые части. `join()` добавляет ровно свой разделитель и не восстанавливает прежние пробелы.


```python
text = "  кот   пёс  "
print(text.split())
print(text.split(" "))
print(" / ".join(text.split()))
```

## 6. Собираем части: `join()`

`join()` работает в обратную сторону: соединяет части списка в одну строку. Разделитель пишется перед точкой: `", ".join(список)`.

![Как split делит строку, а join собирает её обратно](assets/split_join.png)


```python
drinks = ["чай", "сок", "вода"]
result = ", ".join(drinks)
print(result)
```

У разделителя может быть другой вид. Например, добавим между словами вертикальную черту:


```python
names = ["Аня", "Борис", "Вера"]
print(" | ".join(names))
```

**Важно:** `join()` соединяет строки. Если в списке числа, сначала преобразуй каждое число в строку.


```python
numbers = [3, 8, 12]
number_parts = [str(number) for number in numbers]
number_text = ", ".join(number_parts)
print(number_text)
```

## 7. Проверяем содержимое строки

Некоторые методы отвечают, состоит ли строка из букв или цифр. Они тоже возвращают `True` или `False`.


```python
print("Python".isalpha())
print("2026".isdigit())
print("Python 3".isalnum())
print("   ".isspace())
```

Пробел — отдельный символ, поэтому `"Python 3".isalnum()` даёт `False`. У `"-15".isdigit()` тоже результат `False`: знак минуса не цифра.

Есть тонкость Unicode: `"²".isdigit()` даёт `True`, но `int("²")` не работает. `isdigit()` не гарантирует успешное преобразование.

В следующем примере правило намеренно узкое: принимаем только цифры ASCII от 0 до 9. `isascii()` проверяет набор символов, `isdecimal()` — непустую запись десятичными цифрами. Для общего целого ввода со знаком используй `try: int(text)` и `except ValueError` после урока об ошибках.


```python
user_text = input("Введи неотрицательное целое число: ").strip()
if user_text.isascii() and user_text.isdecimal():
    number = int(user_text)
    print(f"Через один год будет {number + 1}")
else:
    print("Нужна непустая запись цифрами от 0 до 9 без знака.")
```
