## 1. Параметры и аргументы

В объявлении функции имена внутри скобок называются **параметрами**. Значения, которые мы передаём при вызове, — **аргументами**.

Вызов `add_numbers(3, 4)` передаёт аргументы `3` и `4` параметрам `first` и `second`.


```python
def add_numbers(first, second):
    return first + second

result = add_numbers(3, 4)
print(result)
```

![Как аргументы поступают в функцию, а return возвращает результат](assets/function_flow.png)

При позиционном вызове Python сопоставляет значения и параметры слева направо.


```python
def introduce(name, city):
    print(f"Меня зовут {name}. Я живу в городе {city}.")

introduce("Аня", "Казань")
```

## 2. Именованные аргументы

Можно указать имя параметра при вызове. Тогда порядок аргументов можно изменить, а вызов легче читать.


```python
def introduce(name, city):
    print(f"Меня зовут {name}. Я живу в городе {city}.")

introduce(city="Тула", name="Борис")
```

В одной функции можно сочетать позиционные и именованные аргументы. Позиционные аргументы указывай сначала.


```python
def make_card(name, hobby, city):
    return f"{name} из города {city} любит {hobby}."

card = make_card("Вера", city="Самара", hobby="рисовать")
print(card)
```

## 3. Значения по умолчанию

Если параметр часто имеет одно и то же значение, можно задать значение по умолчанию. Тогда при вызове его можно не указывать.


```python
def greet(name, greeting="Привет"):
    return f"{greeting}, {name}!"

print(greet("Аня"))
print(greet("Борис", "Добро пожаловать"))
```

Параметры со значениями по умолчанию записывают после обязательных параметров.


```python
def print_ticket(name, seat="без места"):
    print(f"Билет для {name}: {seat}")

print_ticket("Вера")
print_ticket("Глеб", seat="ряд 2, место 5")
```

### Важный нюанс: не используй изменяемый список как значение по умолчанию

Python создаёт значение по умолчанию один раз — в момент объявления функции. Если это список и функция его меняет, список сохранится для следующих вызовов.

Пока просто запомни безопасный шаблон: укажи `None`, а пустой список создай внутри функции.


```python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item("карандаш"))
print(add_item("тетрадь"))
```

Если список передан явно, функция добавит элемент в этот список. Это полезно, если такое изменение задумано.


```python
school_items = ["ручка"]
result = add_item("линейка", school_items)

print(result)
print(school_items)
```

## 4. Несколько аргументов: `*args` и `**kwargs`

Иногда заранее неизвестно, сколько значений передадут функции.

- Параметр `*args` собирает дополнительные позиционные аргументы в кортеж.
- Параметр `**kwargs` собирает дополнительные именованные аргументы в словарь.

Названия `args` и `kwargs` — обычные имена. Важны звёздочки `*` и `**`.


```python
def show_scores(*scores):
    print(scores)
    for score in scores:
        print("Оценка:", score)

show_scores(5, 4, 5)
```

Аргументы после `**` удобно использовать, когда набор именованных значений заранее неизвестен.


```python
def show_profile(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

show_profile(name="Аня", city="Казань", hobby="шахматы")
```

### Порядок параметров

Если в одной функции нужны обычные параметры, `*args` и `**kwargs`, располагай их именно в таком порядке.


```python
def describe_trip(destination, *companions, **details):
    print("Куда:", destination)
    print("С кем:", companions)
    print("Подробности:", details)

describe_trip("Тула", "Аня", "Борис", days=2, transport="поезд")
```

Пока не добавляй `*args` и `**kwargs` в каждую функцию. Используй их, когда действительно нужно принимать произвольное количество аргументов.

## 5. Область видимости переменных

Переменная, созданная внутри функции, называется **локальной**. Она доступна внутри этой функции, но не снаружи.


```python
def show_message():
    message = "Привет!"
    print(message)

show_message()
# print(message)  # Переменная message существует только внутри функции
```

Параметры тоже локальные переменные функции. Вызов функции создаёт для них значения.


```python
def double(number):
    result = number * 2
    return result

first_result = double(3)
second_result = double(10)
print(first_result)
print(second_result)
```

Если переменная внутри функции и переменная снаружи имеют одинаковое имя, внутренняя переменная не изменяет внешнюю.


```python
score = 100

def show_score():
    score = 5
    print("Внутри функции:", score)

show_score()
print("Снаружи функции:", score)
```

Обычно функции лучше передавать нужные значения через параметры, а результат получать через `return`. Так понятно, откуда функция берёт данные и что она возвращает.


### Ранний `return`

`return` сразу завершает функцию. Это удобно, если нужно быстро обработать отдельный случай.


```python
def describe_number(number):
    if number == 0:
        return "Это ноль"

    if number > 0:
        return "Это положительное число"

    return "Это отрицательное число"

print(describe_number(0))
print(describe_number(-3))
```

## 6. Документация и аннотации типов

**Docstring** — короткое описание функции. Его пишут первой строкой внутри тела функции в тройных кавычках.


```python
def calculate_area(width, height):
    """Возвращает площадь прямоугольника."""
    return width * height

print(calculate_area(5, 3))
print(calculate_area.__doc__)
```

Аннотации типов подсказывают читателю, какие значения ожидаются и что возвращает функция. После двоеточия указывают тип параметра, а после `->` — тип результата.


```python
def calculate_area(width: int, height: int) -> int:
    """Возвращает площадь прямоугольника со сторонами в целых числах."""
    return width * height

area = calculate_area(5, 3)
print(area)
```

Аннотации помогают читать код и работают вместе с инструментами проверки. **Python сам по себе не запрещает** передать значение другого типа только из-за аннотации.


```python
def repeat_word(word: str, count: int) -> str:
    return word * count

print(repeat_word("ха", 3))
```

## Короткое повторение

- **Параметры** записываются в объявлении функции, **аргументы** передаются при вызове.
- Именованные аргументы делают вызовы понятнее.
- Значения по умолчанию позволяют не указывать необязательные параметры.
- `*args` собирает позиционные аргументы в кортеж, `**kwargs` — именованные аргументы в словарь.
- Локальная переменная доступна только внутри функции.
- `return` завершает функцию и отдаёт результат.
- Docstring описывает функцию, аннотации типов помогают читать код, но сами не проверяют типы во время выполнения.
