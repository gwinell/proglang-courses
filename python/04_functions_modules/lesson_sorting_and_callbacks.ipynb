# Сортировка по признаку и функции как аргументы

Обычная сортировка умеет сравнивать строки и числа. Но в списке записей нужно объяснить Python, **по какому признаку** сравнивать элементы. Для этого `sorted` и метод `.sort()` принимают функцию-параметр `key`.

## 1. Что именно сортирует Python

`sorted(последовательность)` создаёт и возвращает новый список. Метод `список.sort()` меняет сам список и возвращает `None`. Оба варианта по умолчанию идут по возрастанию.

Для простых чисел и строк порядок очевиден. Для словаря запись целиком сравнивать не нужно: сначала извлеки из неё значение, которое будет ключом сортировки.

```python
names = ["Мария", "Илья", "Александр"]
copy_sorted = sorted(names)
print(copy_sorted)
print(names)

names.sort()
print(names)
```

## 2. Готовая функция `key=len`

Параметр `key` принимает функцию. Python вызывает её для каждого элемента, получает ключ, а элементы упорядочивает по этим ключам. Запись `key=len` передаёт саму функцию `len`; запись `key=len()` попыталась бы вызвать `len` прямо сейчас без аргумента.

Например, если ключ — длина слова, сортируется не алфавитная последовательность, а числа 2, 5, 9. Равные ключи сохраняют исходный взаимный порядок: сортировка в Python стабильна.

```python
words = ["груша", "кот", "слон", "ёж"]
by_length = sorted(words, key=len)
print(by_length)
print(sorted(words, key=len, reverse=True))
```

## 3. Именованная функция-ключ

Если правило требует пояснения или используется несколько раз, запиши его обычной функцией. Она должна принять один элемент и вернуть значение, по которому его сравнивают.

Список кортежей хранит имя и результат. Ключ `score_of` получает один кортеж и возвращает его второй элемент. Сама функция передаётся без скобок.

```python
results = [("Лена", 8), ("Олег", 5), ("Маша", 10)]

def score_of(result):
    return result[1]

print(sorted(results, key=score_of))
```

## 4. `lambda`: короткая функция для одного выражения

Когда правило маленькое и используется один раз, можно записать его прямо в `key`: `lambda result: result[1]`. Это функция без имени: после двоеточия стоит **одно выражение**, значение которого и возвращается.

Осваивай сначала именованную функцию и только потом сокращай запись. Если ключ становится длинным, содержит несколько шагов или нуждается в пояснении, оставь обычный `def`.

```python
print(sorted(results, key=lambda result: result[1], reverse=True))
```

## 5. Несколько полей и равные значения

Ключом может быть кортеж. Python сравнивает поля слева направо: сначала баллы, затем имя, если баллы равны. В этом примере больший балл должен идти первым, а при равных баллах имена — по алфавиту. Нельзя просто добавить `reverse=True`: он развернул бы порядок и имён. Составь ключ, где первый критерий имеет обратный знак, а второй остаётся обычным.

Для общих данных выбирай направление по каждому признаку явно. Стабильность позволяет выполнять многоступенчатую сортировку: сначала по вторичному полю, затем стабильной сортировкой по главному.

```python
players = [
    {"name": "Зоя", "score": 12},
    {"name": "Ада", "score": 18},
    {"name": "Марк", "score": 18},
    {"name": "Борис", "score": 12},
]

ranked = sorted(players, key=lambda player: (-player["score"], player["name"]))
print(ranked)

# Сначала вторичный критерий, затем стабильная сортировка по главному.
by_name = sorted(players, key=lambda player: player["name"])
by_score_then_name = sorted(by_name, key=lambda player: player["score"], reverse=True)
print(by_score_then_name)
```
