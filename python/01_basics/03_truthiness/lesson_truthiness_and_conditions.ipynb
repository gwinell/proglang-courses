# Истинность значений и короткое вычисление условий

Перед тем как использовать списки в условиях, разберём, почему `if 0:` и `if "":` выполняют ветку `else`. Новые идеи: истинность простых значений и короткое вычисление `and`/`or`. Здесь достаточно чисел и строк.

## Результаты урока

После урока ты сможешь: предсказать ветку для нуля и непустого значения; заменить проверку равенства нулю на условие истинности, если это точно соответствует задаче; объяснить порядок вычисления `and` и `or`; выбрать явную запись, если она понятнее.


## 1. Ноль и ненулевое число

В условии Python трактует число `0` как ложь, а любое другое число — как истину. Это не меняет значение переменной: Python лишь решает, выполнять ли тело `if`.


```python
number = 0

if number:
    print("Число не ноль")
else:
    print("Число равно нулю")
```

**Предскажи до запуска:** что появится, если заменить `0` на `-2`? А на `3`? Измени только число и проверь оба случая. Отрицательное число тоже истинно, потому что оно не равно нулю.

## 2. Пустая и непустая строка

Пустая строка `""` ложна. Строка хотя бы с одним символом истинна. Строка `"0"` — непустая, поэтому истинна, хотя число `0` ложно. Сначала строка и число — разные типы.


```python
text = ""

if text:
    print("Текст есть")
else:
    print("Текст пустой")
```

**Измени:** проверь `"Python"`, один пробел `" "` и `"0"`. Пробел — символ, поэтому строка из пробела не пуста. Если в задаче пробелы вокруг ответа не должны считаться текстом, это отдельное правило: сначала нужно убрать пробелы отдельным правилом. Метод `strip()` мы изучим позже.

## 3. Когда короткая проверка подходит

Если нужно спросить «введено ли ненулевое количество билетов?», условие `if tickets:` подходит. Если нужно спросить именно «равно ли значение нулю?», запись `if tickets == 0:` обычно понятнее. Выбирай запись, которая прямо выражает вопрос задачи.


```python
tickets = 0

if tickets == 0:
    print("Билетов пока нет")

if tickets:
    print("Можно продолжать оформление")
```

Оба условия могут быть полезны в разных местах программы. Не считай `if value:` универсальной заменой любой проверки: например, отрицательный остаток тоже будет истинным.

## 4. `and`: вторая часть нужна только если первая истинна

Сначала Python вычисляет левую часть. Если она ложна, результат `and` уже не может быть истинным, поэтому правая часть не вычисляется. Это короткое вычисление. Благодаря этому справа можно поставить проверку, безопасную только при определённом условии слева.


```python
age = 12

if age >= 18 and age <= 120:
    print("Возраст в указанном диапазоне")
else:
    print("Возраст вне диапазона")
```

**Проследи порядок:** при `age = 12` первая часть ложна. Правую часть Python не проверяет. Измени значение на `25`, затем на `121`. Объясни результаты до запуска.

Пока не используй справа выражение, которое может вызвать ошибку, если ты ещё не умеешь уверенно объяснить, почему оно будет или не будет вычислено.

## 5. `or`: вторая часть нужна только если первая ложна

Для `or` достаточно одной истинной части. Если левая часть истинна, правая уже не изменит ответ и не выполняется.


```python
answer = ""

if answer == "да" or answer == "д":
    print("Продолжаем")
else:
    print("Ответ не распознан")
```

Измени `answer` на `"д"`, `"да"` и `"нет"`. Затем поменяй части условия местами. Результат останется тем же, но порядок вычисления изменится.

Условия соединяй скобками, если скобки делают замысел яснее. В этой теме не нужно запоминать таблицу приоритетов наизусть: при сомнении поставь скобки.

### ДЗ-3.1. Пустое сообщение · обязательная

**Сюжет.** Робот принимает текст сообщения. Даже один символ означает, что сообщение поступило.

**Условие.** Проверьте, пуста ли введённая строка. Пробелы и цифры считаются обычными символами.

**Формат данных**

| Входные данные | Выходные данные |
|---|---|
| Одна строка `message`, которая может быть пустой. | `Сообщение получено`, если строка непустая; иначе `Нет сообщения`. |

**Ограничения.** От 0 до 80 символов. Не удаляйте пробелы и не преобразовывайте текст в число.

**Примеры**

| Входные данные | Выходные данные |
|---|---|
| <em>Пустая строка (Enter)</em> | <pre>Нет сообщения</pre> |
| <pre>0</pre> | <pre>Сообщение получено</pre> |

**Пояснение.** Первый ввод — одно нажатие Enter без символов. Строка «0» содержит символ и потому не пуста.

### ДЗ-3.3. Пропуск с именем · обязательная

**Условие.** Разрешите вход, если имя непустое, а числовой код равен 2026. Иначе откажите.

**Формат данных**

| Входные данные | Выходные данные |

### ДЗ-3.4. Ответ на приглашение · по выбору

**Сюжет.** Бот принимает два варианта согласия на участие в игре.

**Условие.** Считайте согласием только ответы `да` и `д`. Любая другая строка означает отказ.

**Формат данных**

| Входные данные | Выходные данные |

### ДЗ-3.5. Вода для команды · по выбору

**Сюжет.** Команда рассчитывает, хватит ли воды из общего запаса: каждому требуется не меньше двух литров.

**Условие.** Если участников больше нуля и запаса хватит каждому на два литра, подтвердите готовность. Если участников нет или воды не хватает, напечатайте отказ.

**Формат данных**

| Входные данные | Выходные данные |
|---|---|
| Две строки: целое число участников `people`, затем целое количество литров `water`. | `Можно отправляться` или `Подготовка не завершена`. |

**Ограничения.** 0 ≤ people ≤ 100; 0 ≤ water ≤ 1000. Делить на ноль нельзя: случай без участников должен дать отказ.

**Примеры**

| Входные данные | Выходные данные |
|---|---|
| <pre>3<br>6</pre> | <pre>Можно отправляться</pre> |
| <pre>3<br>5</pre> | <pre>Подготовка не завершена</pre> |
| <pre>0<br>6</pre> | <pre>Подготовка не завершена</pre> |

**Пояснение.** Ровно двух литров на человека достаточно. Запас воды не разрешает отправление команды из нуля участников.
