# Строки и списки в циклах

В этом конспекте учимся обрабатывать строки и списки с помощью циклов. Разберём два способа: брать элементы по очереди или идти по индексам через range().

Сначала идут объяснения простыми словами. В конце есть отдельный раздел с точными формулировками «Как сказать в колледже».

Домашних заданий в этом конспекте нет.

## Что ты узнаешь

- как цикл for проходит по символам строки;
- как for проходит по элементам списка;
- чем элемент отличается от индекса;
- как range(len(...)) помогает обратиться к каждому элементу по номеру;
- как использовать условие внутри цикла;
- почему строка и список по-разному меняются.

## 1. Строка и список: что у них общего

Строка — это текст из символов. Например, слово КОТ состоит из символов К, О, Т.

Список — это несколько значений, записанных по порядку. Например, список животных состоит из слов кот, пёс, ёж.

И у символов строки, и у элементов списка есть порядок и индексы. Счёт индексов начинается с нуля:

| Индекс | Строка КОТ | Список животных |
|---:|:---:|---|
| 0 | К | кот |
| 1 | О | пёс |
| 2 | Т | ёж |

![Строка и список разбиты на элементы; показаны индексы и последовательная обработка каждого элемента](https://raw.githubusercontent.com/gwinell/proglang-courses/main/python/special/strings_lists_loops_imagegen.png)

В кружках на схеме показан номер шага цикла. Цифры под буквами и словами — индексы. Например, на первом шаге цикла обрабатывается элемент с индексом 0.

## 2. Цикл for берёт элементы по очереди

Простыми словами: цикл берёт одно значение, выполняет с ним команды, затем берёт следующее. Он повторяется, пока значения не закончатся.

В записи for letter in word переменная letter по очереди получает каждый символ строки word.

```python
word = "КОТ"

for letter in word:
    print(letter)
```

Программа напечатает:

    К
    О
    Т

Сначала в letter будет буква К, потом О, затем Т. После последнего символа цикл закончится.

## 3. Такой же способ работает со списком

Если в источнике список, переменная цикла получает по очереди элементы этого списка.

```python
animals = ["кот", "пёс", "ёж"]

for animal in animals:
    print("Я вижу:", animal)
```

Вывод:

    Я вижу: кот
    Я вижу: пёс
    Я вижу: ёж

В каждом круге в animal лежит один элемент. Цикл сам переходит к следующему элементу.

## 4. Элемент и индекс — разные вещи

В примере for animal in animals переменная animal получает само значение, например слово «пёс». Номер ячейки в ней не хранится.

Индекс — это номер позиции. Первый индекс равен 0, следующий — 1 и так далее.

Если нужно напечатать номер позиции или получить значение именно по его индексу, используем range(len(...)).

## 5. range(len(...)): проходим по индексам

len(word) сообщает длину строки. У строки КОТ длина равна 3.

range(len(word)) даёт индексы 0, 1, 2. На каждом круге word[index] берёт символ на текущем индексе.

У списка можно точно так же получить длину и элемент по индексу.

![Индексы 0, 1 и 2 по очереди выбирают символ из строки word и элемент из списка animals](https://raw.githubusercontent.com/gwinell/proglang-courses/main/python/special/strings_lists_indexes_imagegen.png)

```python
word = "КОТ"

for index in range(len(word)):
    print("Индекс:", index, "Символ:", word[index])
```

Вывод:

    Индекс: 0 Символ: К
    Индекс: 1 Символ: О
    Индекс: 2 Символ: Т

Здесь index меняется: сначала 0, потом 1, затем 2. Запись word[index] каждый раз берёт символ на этой позиции.

```python
animals = ["кот", "пёс", "ёж"]

for index in range(len(animals)):
    print("Индекс:", index, "Элемент:", animals[index])
```

Результат:

    Индекс: 0 Элемент: кот
    Индекс: 1 Элемент: пёс
    Индекс: 2 Элемент: ёж

Выбирай перебор значений, если номер позиции не нужен. Используй индексы, если нужно показать номер или заменить элемент списка.

## 6. Условие внутри цикла

Условие можно проверять отдельно для каждого текущего значения. В строке КОТИК оставим только гласные О и И. В списке оставим только чётные числа. Новые значения будем собирать отдельно от исходных.

![Схема отбора: цикл проверяет символы строки и числа списка, а подходящие значения собирает в новые результаты](https://raw.githubusercontent.com/gwinell/proglang-courses/main/python/special/strings_lists_filter_imagegen.png)

```python
word = "КОТИК"
vowels = ""

for letter in word:
    if letter == "О" or letter == "И":
        vowels = vowels + letter

print(vowels)
```

Цикл проверит буквы К, О, Т, И, К. К строке vowels добавятся только О и И.

Вывод: ОИ.

Строку нельзя заменить по отдельному индексу, поэтому мы собираем новую строку в переменной vowels.

```python
numbers = [2, 5, 8, 3]
even_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)

print(even_numbers)
```

Проверка number % 2 == 0 спрашивает, делится ли число на 2 без остатка. Если да, append(number) добавляет число в конец нового списка.

Вывод: [2, 8].

Исходный список numbers не меняется. Подходящие элементы попадают в новый список even_numbers.

## 7. Список можно менять по индексу

Список изменяемый. Можно записать новое значение в выбранную ячейку. Например, добавим к каждому слову восклицательный знак. Для этого понадобятся индексы, потому что мы заменяем значения в исходном списке.

```python
animals = ["кот", "пёс", "ёж"]

for index in range(len(animals)):
    animals[index] = animals[index] + "!"

print(animals)
```

Вывод:

    ['кот!', 'пёс!', 'ёж!']

На каждом шаге animals[index] указывает на текущую ячейку. Новое слово записывается в эту ячейку.

## 8. Строку нельзя изменить по индексу

Строка и список ведут себя по-разному. Список можно заменить по индексу, а символ существующей строки — нельзя.

Чтобы получить изменённый текст, создай новую строку и добавляй в неё нужные символы в цикле. Например, собери из слова только гласные — это показано выше.

## 9. Посчитаем подходящие символы

Перед циклом создай переменную count со значением 0. Если текущий символ подходит под условие, увеличь счётчик на один.

```python
word = "КОТИК"
count = 0

for letter in word:
    if letter == "О" or letter == "И":
        count = count + 1

print("Гласных:", count)
```

Программа проверяет каждую букву, находит две гласные и печатает:

    Гласных: 2

Если задача просит только количество, отдельный список или строка результата не обязательны.

## 10. Частые ошибки

- В for animal in animals переменная animal хранит слово, а не номер. Чтобы получить номер, используй range(len(animals)).
- Если длина списка равна 3, индексы — 0, 1, 2. Индекса 3 нет. range(len(animals)) даёт только существующие индексы.
- Перед append() создай пустой список. Перед добавлением символов создай пустую строку.
- Строка не меняется по отдельным символам. Собирай нужный текст в новой строковой переменной.
- Не удаляй элементы из списка, по которому сейчас идёт цикл: после удаления элементы сдвинутся. Для отбора создай новый список.

## Короткое повторение простыми словами

- Строка состоит из символов, список — из элементов.
- В строке и списке значения расположены по порядку, индексы начинаются с 0.
- for letter in word перебирает символы строки.
- for item in items перебирает элементы списка.
- range(len(word)) или range(len(items)) даёт индексы.
- word[index] или items[index] берёт значение по индексу.
- Список можно менять по индексу; строку так менять нельзя.
- if помогает обработать только подходящие значения.

# Как сказать в колледже: термины

**Строка (str)** — неизменяемая последовательность символов.

**Список (list)** — упорядоченная изменяемая последовательность объектов Python.

**Элемент** — одно значение, хранящееся в списке. **Символ** — отдельная единица текста в строке.

**Индекс** — целочисленная позиция элемента в последовательности. Индексация Python начинается с нуля.

**Итерация** — один проход тела цикла для очередного значения. **Итерирование** — последовательное получение значений из строки или списка.

**Цикл for** выполняет блок команд для каждого значения итерируемого объекта. Строка и список являются итерируемыми объектами.

**len()** возвращает количество элементов последовательности. Для строки это количество символов, для списка — количество элементов.

**range()** создаёт последовательность целых чисел, которую часто используют для перебора индексов.

**Фильтрация** — отбор значений, удовлетворяющих заданному условию.

**Аккумулятор** — переменная, в которой накапливается результат: например, счётчик, новая строка или новый список.

Пример ответа: «Строка — неизменяемая последовательность символов. Цикл for позволяет последовательно обработать каждый символ. Если нужен индекс, можно пройти по диапазону range(len(word)) и получить символ выражением word[index]».

Пример ответа про список: «Список — изменяемая упорядоченная последовательность. Цикл for перебирает его элементы; если нужно изменить элемент на месте, можно пройти по индексам и выполнить присваивание items[index] = новое_значение».
