# Дополнительный урок. `break`, `continue`, цикл `for` и `range()`

Раньше мы учились повторять команды с помощью `while` и проверять условия. Теперь разберём две команды, которые управляют циклом, а ещё научимся работать с циклом `for`.

## 1. `break` и `continue` в цикле `while`

Вспомним: `while` повторяет команды, пока его условие верно.

Иногда нужно изменить обычный ход цикла:

- `break` сразу заканчивает весь цикл.
- `continue` пропускает остаток только этого повтора. Затем `while` проверяет условие снова.

Запомни коротко: **`break` — закончить; `continue` — пропустить один повтор.**

````python
number = 0

while number < 6:
    number = number + 1

    if number == 2:
        continue

    if number == 5:
        break

    print(number)

print("После цикла:", number)
````

### Разберём пример по шагам

1. Сначала `number` равно 0. Проверка `number < 6` верна, поэтому цикл начинается.
2. Внутри цикла прибавляем 1. Теперь `number` равно 1. Число печатается.
3. В следующий раз `number` становится 2. Выполняется `continue`, поэтому печать пропускается. Цикл возвращается к проверке условия.
4. Для 3 и 4 печать выполняется.
5. Когда `number` становится 5, срабатывает `break`. Цикл сразу заканчивается — число 5 не печатается.
6. После цикла печатается значение `number`.

**Весь вывод:**

```text
1
3
4
После цикла: 5
```

В цикле `while` меняй число **до** `continue`. Иначе можно случайно пропустить изменение, нужное для завершения цикла, и он будет повторяться бесконечно.

## 2. Что делает цикл `for`

`for` повторяет команды для значений, которые получает по очереди. Например, ему можно дать числа 1, 2 и 3.

```python
for number in range(1, 4):
    print(number)
```

Вот как прочитать эту запись:

- `for` — начинаем повторять команды;
- `number` — переменная, куда цикл будет по очереди класть очередное число;
- `in` значит «из»;
- `range(1, 4)` задаёт числа 1, 2, 3;
- двоеточие `:` отмечает начало команд цикла;
- отступ перед `print(number)` показывает, что команда повторяется.

**Что делает Python:** сначала кладёт в `number` число 1 и выполняет `print()`, потом кладёт 2 и выполняет `print()`, затем делает то же с числом 3. Числа заканчиваются — цикл заканчивается. Число 4 не берётся.

На схеме каждое число проходит через тело цикла по очереди. После последнего числа программа выходит из цикла.

![Схема цикла for: числа 1, 2 и 3 по очереди проходят через print](for_loop_imagegen.png)

````python
for number in range(1, 4):
    print("Получили число:", number)

print("Числа закончились")
````

**Пример вывода:**

```text
Получили число: 1
Получили число: 2
Получили число: 3
Числа закончились
```

Переменная цикла меняется сама: 1, затем 2, затем 3. Не нужно писать `number = number + 1`.

## 3. Как работает `range()`

`range()` задаёт числа для цикла. В него можно передать одно, два или три целых числа.

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

Слова `start`, `stop` и `step` — подсказки о месте для числа:

- `start` — с какого числа начать;
- `stop` — перед каким числом остановиться;
- `step` — на сколько менять число при переходе к следующему.

**Общее правило:** `stop` не включается. Цикл заканчивается, когда подошёл к границе остановки.

### Случай 1. Одно число: `range(stop)`

Если указано только одно число, отсчёт начинается с нуля.

Например, `range(5)` даёт числа `0, 1, 2, 3, 4`. Число 5 — это граница: перед ним остановимся.

| Запись | Какие числа придут в цикл | Сколько повторов |
|---|---|---:|
| `range(5)` | 0, 1, 2, 3, 4 | 5 |
| `range(1)` | 0 | 1 |
| `range(0)` | ни одного числа | 0 |

Значит, `range(5)` повторяет действие пять раз, но счётчик начинается с 0.

````python
for number in range(5):
    print(number)
````

**Вывод:**

```text
0
1
2
3
4
```

Python начинает с 0 и прибавляет 1. До 5 он не доходит: 5 — граница остановки.

### Случай 2. Два числа: `range(start, stop)`

Первое число — начало. Второе число — граница остановки. Если шаг не указан, Python каждый раз прибавляет 1.

`range(2, 6)` читаем так: начать с 2, идти вверх по одному и остановиться перед 6.

```text
2 → 3 → 4 → 5 → остановка перед 6
```

| Запись | Какие числа придут в цикл | Почему |
|---|---|---|
| `range(2, 6)` | 2, 3, 4, 5 | начинаем с 2, не включаем 6 |
| `range(1, 6)` | 1, 2, 3, 4, 5 | начинаем с 1, не включаем 6 |
| `range(4, 4)` | чисел нет | начинаем прямо на границе |
| `range(5, 2)` | чисел нет | обычный шаг идёт вверх, но начало уже больше границы |

#### Как вывести числа от 1 до 5 включительно?

Пиши `range(1, 6)`. Число 6 не попадёт в цикл, значит последним будет 5.

Если конечное число хранится в `n`, чтобы включить его, пиши `range(1, n + 1)`. Например, если `n` равно 4, `range(1, n + 1)` превращается в `range(1, 5)` и даёт 1, 2, 3, 4.

````python
n = 5

for number in range(1, n + 1):
    print(number)
````

**Вывод при `n = 5`:**

```text
1
2
3
4
5
```

`n + 1` нужно только для границы остановки. Само число `n` при этом попадает в ряд.

### Случай 3. Три числа: `range(start, stop, step)`

Третье число — шаг. Шаг говорит, на сколько число меняется при следующем повторе.

#### Считаем вверх

`range(1, 8, 2)` означает: начать с 1, прибавлять 2, остановиться перед 8.

```text
1 → 3 → 5 → 7 → остановка
```

Следующим было бы 9. Оно уже прошло бы границу 8, поэтому ряд заканчивается. Число 8 тоже не включается.

| Запись | Числа | Что делаем |
|---|---|---|
| `range(1, 8, 2)` | 1, 3, 5, 7 | прибавляем 2 |
| `range(2, 11, 2)` | 2, 4, 6, 8, 10 | прибавляем 2 |
| `range(0, 10, 3)` | 0, 3, 6, 9 | прибавляем 3 |

#### Считаем вниз

Чтобы идти вниз, шаг должен быть отрицательным. `range(5, 0, -1)` означает: начать с 5, вычитать 1, остановиться перед 0.

```text
5 → 4 → 3 → 2 → 1 → остановка перед 0
```

| Запись | Числа | Что делаем |
|---|---|---|
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 | вычитаем 1 |
| `range(10, 0, -2)` | 10, 8, 6, 4, 2 | вычитаем 2 |
| `range(7, 2, -2)` | 7, 5, 3 | вычитаем 2, останавливаемся перед 2 |

**Правила шага:** положительный шаг ведёт вверх, отрицательный — вниз. Шаг не должен быть 0. Направление должно вести от начала к границе: `range(2, 5, -1)` не даст чисел, потому что шаг ведёт вниз, хотя нужно идти вверх.

Число остановки `stop` не включается и в этой записи.

````python
print("Считаем вверх:")
for number in range(1, 8, 2):
    print(number)

print("Считаем вниз:")
for number in range(5, 0, -1):
    print(number)
````

**Вывод:**

```text
Считаем вверх:
1
3
5
7
Считаем вниз:
5
4
3
2
1
```

Перед запуском `range()` проверь четыре вещи: число начала, границу остановки, знак шага и само значение шага.

## 4. Условия внутри `for`

Внутри цикла можно ставить `if`, `elif` и `else`. На каждом повторе проверяется текущее число.

В этом примере проверяем числа от 1 до 5. Если число делится на 2 без остатка, печатаем «чётное». Иначе печатаем «нечётное».

`%` показывает остаток от деления. Например, `4 % 2` равно 0: число 4 делится на 2 без остатка.

Отступы показывают, что весь `if / else` находится внутри цикла, а `print()` — внутри своей ветки условия.

````python
for number in range(1, 6):
    if number % 2 == 0:
        print(number, "— чётное")
    else:
        print(number, "— нечётное")
````

**Вывод:**

```text
1 — нечётное
2 — чётное
3 — нечётное
4 — чётное
5 — нечётное
```

Python проверяет условие отдельно для каждого числа.

## 5. `break` в цикле `for`

`break` означает: «закончить цикл сейчас».

Цикл в примере мог бы взять числа от 1 до 5. Но перед печатью программа проверяет, равно ли текущее число 4.

- Для 1, 2 и 3 проверка ложная — числа печатаются.
- Для 4 проверка верная — выполняется `break`.
- Число 4 не печатается, потому что команда `break` стоит перед `print()`.
- До числа 5 цикл уже не дойдёт.

Команды после цикла всё равно будут выполнены.

````python
for number in range(1, 6):
    if number == 4:
        break
    print(number)

print("После цикла")
````

**Вывод:**

```text
1
2
3
После цикла
```

`break` закончил цикл, а не всю программу.

## 6. `continue` в цикле `for`

`continue` означает: «пропустить остаток только этого повтора».

В примере число 3 не печатается. Когда `number` равно 3, команда `continue` перескакивает через `print(number)`. Затем цикл `for` сам берёт следующее число — 4.

В `for` не нужно вручную увеличивать `number`: следующее значение берётся из `range()`.

````python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)

print("Цикл завершён")
````

**Вывод:**

```text
1
2
4
5
Цикл завершён
```

Цикл продолжил работу после числа 3. Это главное отличие `continue` от `break`.

### Сравним команды на схеме

В верхней части схемы число 3 пропускается, но цикл переходит к 4. Это `continue`.

В нижней части после 4 срабатывает `break`. Цикл заканчивается, поэтому число 5 больше не проверяется.

![Схема: continue пропускает одно число, break завершает цикл](for_break_continue_imagegen.png)

| Команда | Что сделать | Что будет потом |
|---|---|---|
| `continue` | пропустить текущий повтор | перейти к следующему числу |
| `break` | закончить цикл | выполнять команды после цикла |

## Термины: два варианта

### 1. Чтобы понять

- **Цикл `for`** — команда, которая по очереди берёт значения и повторяет другие команды для каждого.
- **Переменная цикла** — переменная, в которую на каждом повторе попадает следующее значение. Здесь это `number`.
- **`range()`** — способ задать ряд целых чисел для цикла.
- **Начало `start`** — первое число ряда.
- **Остановка `stop`** — граница, перед которой ряд заканчивается. Само это число не включается.
- **Шаг `step`** — на сколько меняется число при следующем повторе.
- **Тело цикла** — команды с отступом, которые повторяются.
- **Итерация** — один повтор команд цикла.
- **`break`** — команда, которая сразу заканчивает цикл.
- **`continue`** — команда, которая пропускает остаток текущего повтора и переходит к следующему.
- **Условие** — проверка, которая помогает выбрать действие, например `number == 4`.

### 2. Чтобы сказать в колледже

- **Цикл `for`** — конструкция языка Python для последовательного выполнения блока команд для каждого значения последовательности.
- **Переменная цикла** — имя, которому на каждой итерации присваивается очередное значение последовательности.
- **`range()`** — встроенный тип для представления последовательности целых чисел с началом, исключающей верхней границей и шагом.
- **Начало (`start`)** — первое значение последовательности `range`.
- **Верхняя граница (`stop`)** — значение, которое задаёт окончание последовательности, но не включается в неё.
- **Шаг (`step`)** — разность между соседними значениями последовательности.
- **Тело цикла** — блок команд, выполняемый на каждой итерации.
- **Итерация** — один проход выполнения тела цикла.
- **Оператор `break`** — оператор, немедленно завершающий ближайший охватывающий цикл.
- **Оператор `continue`** — оператор, завершающий текущую итерацию и передающий управление следующей итерации ближайшего цикла.
- **Условная конструкция** — конструкция `if / elif / else`, выбирающая ветку выполнения по результату проверки.
