## 1. Что такое тип данных?

Тип данных показывает, с какими данными мы работаем.

- `int` — целое число: `12`.
- `float` — число с точкой: `2.5`.
- `str` — текст: `"Привет"`.
- `bool` — значение `True` или `False`.

`True` значит «да» или «верно». `False` значит «нет» или «неверно».

Важно: текст `"12"` и число `12` имеют разные типы. Кавычки показывают, что это текст.


````python
print(type(12))
print(type(2.5))
print(type("Привет"))
print(type(True))
````

`type()` показывает тип значения.

В ответе Python пишет `<class 'int'>`. Для начала смотрите на последнее слово: `int`, `float`, `str` или `bool`.


### `=` и `==` — разные знаки

Один знак равенства `=` сохраняет значение в переменную.

Два знака `==` сравнивают два значения. Результат сравнения — `True` или `False`.

```python
age = 12       # сохраняем число 12
print(age == 12)  # проверяем: age равно 12?
```


````python
age = 12
print(age == 12)
print(age == 10)
````

## 2. Сравнение чисел

Знаки сравнения:

- `<` — меньше;
- `<=` — меньше или равно;
- `>` — больше;
- `>=` — больше или равно;
- `==` — равно;
- `!=` — не равно.

После сравнения Python получает `True` или `False`.


![Весы показывают сравнение: на одной чаше два предмета, на другой три](comparison_scale.png)

````python
print(5 < 6)
print(5 <= 5)
print(9 > 3)
print(9 >= 12)
print(3 == 3)
print(4 != 3)
````

Прочитаем несколько примеров:

- `5 < 6` — верно, поэтому результат `True`;
- `9 >= 12` — неверно, поэтому результат `False`;
- `4 != 3` спрашивает: «4 не равно 3?» Ответ `True`.


## 3. Тип `bool` и команда `bool()`

Тип `bool` хранит только два значения: `True` и `False`.

Команда `bool(...)` проверяет значение и возвращает `True` или `False`.

Для чисел:

- `0` даёт `False`;
- любое число, кроме нуля, даёт `True`.

Это правило работает и для отрицательных, и для дробных чисел.


````python
print(bool(1))
print(bool(0))
print(bool(1234))
print(bool(-5))
print(bool(2.5))
print(bool(-2.5))
````

У `bool(0)` результат `False`, потому что число равно нулю. У `bool(-5)` результат `True`: число не равно нулю.


### Проверка текста

Для текста правило такое:

- пустой текст `""` даёт `False`;
- любой непустой текст даёт `True`.

Пробел — тоже часть текста. Поэтому `" "` не пустой. Текст `"0"` тоже не пустой.


````python
print(bool(" "))
print(bool("Привет"))
print(bool("0"))
print(bool(""))
````

`bool("0")` даёт `True`, потому что внутри кавычек есть символ `0`. Это текст, а не число ноль.


## 4. Условие `if`

`if` значит «если».

Python проверяет условие после `if`:

- если ответ `True`, выполняет команды с отступом;
- если ответ `False`, пропускает эти команды.

После условия ставится двоеточие `:`. Команды внутри условия пишутся с отступом — обычно четыре пробела.


````python
age = 10
if age < 12:
    print("Вход запрещён")
print("Продолжение программы")
````

В примере `age < 12` верно, потому что `age` равно 10. Поэтому Python выводит `Вход запрещён`.

Последний `print()` не имеет отступа. Он выполняется после условия в любом случае.


## 5. Несколько вариантов: `elif` и `else`

`elif` значит «иначе, если».

`else` значит «иначе».

Python проверяет условия сверху вниз. Он выполняет команды только в первой подходящей части:

1. Сначала проверяет `if`.
2. Если условие неверно, проверяет `elif`.
3. Если все условия неверны, выполняет `else`, если он есть.

Порядок важен: после найденного подходящего условия остальные части этой цепочки уже не выполняются.


![Схема выбора одного пути в условии if, elif, else](if_branch_paths.png)

Посмотрим на пример с возрастом:

- меньше 12 лет — вход запрещён;
- от 12 до 15 лет — вход разрешён с родителем;
- от 16 до 18 лет — вход разрешён до 19:00;
- с 19 лет — вход разрешён.

Эти границы следуют из условий в коде.


````python
age = 15
if age < 12:
    print("Вход запрещён")
elif age < 16:
    print("Вход разрешён с родителем")
elif age <= 18:
    print("Вход разрешён до 19:00")
else:
    print("Вход разрешён")
````

У `age` значение 15.

- `age < 12` — неверно;
- `age < 16` — верно.

Поэтому выполняется вторая ветка. До третьей ветки Python уже не доходит.


### Пример с вводом

`input()` получает текст. Если ответ нужен как число, используйте `int(input(...))`.

После запуска введите возраст и нажмите Enter.


````python
age = int(input("Сколько тебе лет? "))
if age < 12:
    print("Вход запрещён")
elif age < 16:
    print("Вход разрешён с родителем")
elif age <= 18:
    print("Вход разрешён до 19:00")
else:
    print("Вход разрешён")
````

## 6. Условия вместе: `and` и `or`

`and` значит «и». Всё условие верно, только если верны обе его части.

`or` значит «или». Всё условие верно, если верна хотя бы одна часть.

Примеры:

- `a > 0 and b > 0` — оба числа больше нуля;
- `a < 0 or b < 0` — хотя бы одно число меньше нуля.


````python
a = 4
b = 7
print(a > 0 and b > 0)
print(a < 0 or b < 0)
print(a > 0 or b > 0)
````

### Проверим, какие числа положительные

Нужно учесть три случая:

1. Оба числа больше нуля.
2. Хотя бы одно число меньше нуля.
3. Отрицательных чисел нет, и хотя бы одно равно нулю.

Ноль не является ни положительным, ни отрицательным числом. Поэтому для него нужна отдельная ветка `else`.


````python
a = -2
b = 0
if a > 0 and b > 0:
    print("Оба числа положительные")
elif a < 0 or b < 0:
    print("Хотя бы одно число отрицательное")
else:
    print("Отрицательных чисел нет; есть ноль")
````

Здесь `a` равно -2, поэтому условие `a < 0 or b < 0` верно. Ответ `Хотя бы одно число отрицательное` правильный, даже если второе число равно нулю.

Если оба числа равны нулю или одно равно нулю, а второе положительное, программа выведет сообщение из `else`.

## 8. Частые ошибки

- **Поставили `=` вместо `==` в проверке.** `=` сохраняет значение, `==` сравнивает.
- **Забыли двоеточие.** После `if`, `elif` и `else` ставится `:`.
- **Забыли отступ.** Команды внутри ветки пишутся с отступом.
- **Поставили слишком широкое условие первым.** Например, если сначала проверить `age < 16`, ветка для `age < 12` уже никогда не выполнится. Узкую границу нужно проверять первой.
- **Считают текст `"0"` числом ноль.** `"0"` — непустой текст, поэтому `bool("0")` даёт `True`.
- **Забыли про ноль.** Проверка `число > 0` не включает ноль, а проверка `число < 0` тоже его не включает.

При ошибке проверьте знаки, двоеточия, отступы и порядок условий.

## 10. Два варианта определений

У каждого термина есть два объяснения: простое и формальное.

### 1. Чтобы понять

- **Тип данных** — вид данных: число, текст или `True`/`False`.
- **`int`** — тип целых чисел.
- **`float`** — тип чисел с точкой.
- **`str`** — тип текста.
- **`bool`** — тип со значениями `True` и `False`.
- **`type()`** — команда, которая показывает тип значения.
- **Сравнение** — проверка, например «5 больше 3?». Ответ — `True` или `False`.
- **`=`** — знак, который сохраняет значение в переменную.
- **`==`** — знак, который проверяет, равны ли значения.
- **Условие** — вопрос, ответ на который `True` или `False`.
- **`if`** — «если». Команды после него выполняются, когда условие верно.
- **`elif`** — «иначе, если». Проверяет следующий вариант.
- **`else`** — «иначе». Выполняет команды, если предыдущие условия неверны.
- **Отступ** — пробелы перед командой, показывающие, что она относится к условию.
- **`and`** — «и»: должны подойти обе части.
- **`or`** — «или»: должна подойти хотя бы одна часть.
- **`bool()`** — команда, которая проверяет значение и даёт `True` или `False`.
- **Истинность значения** — считается ли значение подходящим для условия. Например, ноль считается ложным, а ненулевое число — истинным.

### 2. Чтобы сказать в колледже

- **Тип данных:** категория объектов, которая определяет их поведение и поддерживаемые операции. В Python встроены, например, типы `int`, `float`, `str` и `bool`.
- **`int`:** встроенный тип целых чисел произвольной точности.
- **`float`:** встроенный тип чисел с плавающей точкой; такие значения обычно представлены приближённо в двоичном формате.
- **`str`:** встроенный тип неизменяемых последовательностей символов Unicode.
- **`bool`:** встроенный логический тип с двумя значениями `True` и `False`; в Python он является подклассом `int`.
- **Функция `type()`:** возвращает тип объекта, переданного аргументом.
- **Операция сравнения:** операция, которая сопоставляет два значения и обычно возвращает логическое значение `True` или `False`.
- **Присваивание `=`:** инструкция, которая вычисляет выражение справа и связывает имя слева с полученным объектом. Это не операция сравнения.
- **Оператор равенства `==`:** оператор сравнения значений на равенство.
- **Условное выражение:** выражение, значение истинности которого проверяется оператором `if`.
- **Инструкция `if`:** составная инструкция управления потоком, которая выполняет тело, если условие истинно.
- **Ветвь `elif`:** дополнительное условие в цепочке `if`; проверяется только если предыдущие условия этой цепочки ложны.
- **Ветвь `else`:** необязательное тело, выполняемое, если ни одно предшествующее условие цепочки `if`/`elif` не истинно.
- **Тело (блок) инструкции:** набор инструкций, принадлежащих ветви. В многострочной форме его задаёт отступ.
- **Логический оператор `and`:** возвращает первый ложный операнд либо последний операнд, если все истинны; вычисляет операнды слева направо и может не вычислять правую часть после ложной левой.
- **Логический оператор `or`:** возвращает первый истинный операнд либо последний операнд, если все ложны; вычисляет операнды слева направо и может не вычислять правую часть после истинной левой.
- **Проверка истинности:** преобразование значения к логическому контексту. Для чисел ложным является ноль, для строк — пустая строка; остальные числа и непустые строки истинны.
- **Функция `bool()`:** возвращает логическое значение, полученное стандартной проверкой истинности своего аргумента.
