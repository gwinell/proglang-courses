# Строки в Python: символы, индексы и срезы

В этом конспекте разберём, как Python хранит текст, как взять один символ по его номеру и как взять часть строки.

Будем использовать примеры и схемы. Здесь нет домашних заданий: кодовые ячейки только показывают работу команд. После каждого примера есть объяснение того, что программа выведет.

## 1. Что такое строка

**Строка** — это текст. Строку записывают в кавычках.

```python
word = "Python"
```

Здесь переменная `word` хранит текст `Python`. Кавычки показывают Python, где текст начинается и где заканчивается. Внутри кавычек можно записать буквы, цифры, пробелы и знаки препинания.

Строку можно вывести на экран командой `print()`.

````python
word = "Python"
print(word)
print("Привет!")
````

Программа напечатает:

```text
Python
Привет!
```

Строка состоит из символов, расположенных по порядку. В слове `Python` шесть символов: `P`, `y`, `t`, `h`, `o`, `n`. У каждого символа есть своё место.

## 2. Строка состоит из символов по порядку

Каждая буква — отдельный символ. Пробел и знак препинания тоже занимают место.

Например, в тексте `"Кот и пёс"` пробелы между словами — тоже символы. Поэтому буква после пробела имеет свой номер, как и любая другая буква.

Python позволяет обратиться к символу по его номеру. Такой номер называется **индексом**.

## 3. Положительные индексы: считаем слева направо

Положительный индекс — это номер символа, если считать слева направо. В Python счёт начинается с нуля:

- первый символ имеет индекс `0`;
- второй — индекс `1`;
- третий — индекс `2`.

Схема показывает строку `PYTHON`: буквы стоят в отдельных ячейках, а над ними указаны индексы.

![Символы PYTHON с положительными индексами 0–5 и отрицательными индексами −6…−1](string_indexes_imagegen.png)

Важно: индекс — это номер места, а не привычный порядковый номер. Первый символ — «первый» для человека, но его индекс в Python равен 0.

````python
word = "PYTHON"

print(word[0])
print(word[1])
print(word[5])
````

В квадратных скобках записан индекс нужного символа.

- `word[0]` берёт символ на месте 0 — `P`.
- `word[1]` берёт символ на месте 1 — `Y`.
- `word[5]` берёт символ на месте 5 — `N`.

**Вывод:**

```text
P
Y
N
```

## 4. Отрицательные индексы: считаем справа налево

Можно считать символы с конца строки. Для этого используют отрицательные индексы:

- индекс `-1` — последний символ;
- индекс `-2` — символ перед последним;
- индекс `-3` — третий символ с конца.

У слова `PYTHON` последний символ — `N`, поэтому `word[-1]` вернёт `N`. Символ перед ним — `O`, поэтому `word[-2]` вернёт `O`.

Посмотри на нижний ряд схемы индексов: он помогает увидеть, как счёт начинается с правого края.

````python
word = "PYTHON"

print(word[-1])
print(word[-2])
print(word[-6])
````

**Вывод:**

```text
N
O
P
```

В этой строке шесть символов. Поэтому `-6` указывает на самый первый символ, а `-1` — на последний.

### Как узнать допустимые индексы

Команда `len()` сообщает, сколько символов в строке:

```python
word = "PYTHON"
print(len(word))
```

В слове `PYTHON` шесть символов, поэтому `len(word)` напечатает `6`.

Если в строке `n` символов, положительные индексы идут от `0` до `n - 1`. Последний индекс равен `len(word) - 1`.

Для `PYTHON`:

- длина — 6;
- индексы слева направо: от 0 до 5;
- индексы справа налево: от -6 до -1.

Чтобы всегда взять последний символ непустой строки, можно написать `word[-1]`. Можно и так: `word[len(word) - 1]`.

````python
word = "PYTHON"
print(len(word))
print(word[len(word) - 1])
````

**Вывод:**

```text
6
N
```

Сначала `len(word)` даёт 6. Индекс последнего символа — на один меньше длины, то есть 5. Выражение `word[len(word) - 1]` поэтому равносильно `word[5]`.

### Что будет, если индекс слишком большой

У строки `PYTHON` есть индексы от 0 до 5. Индекса 6 в ней нет.

Если запросить `word[6]`, Python сообщит об ошибке `IndexError`: такого символа в строке нет. В примерах ниже эту строку кода не запускаем, чтобы ячейка не завершилась ошибкой.

```python
word = "PYTHON"
# word[6]  # такого индекса нет
```

У срезов, которые разберём дальше, правило другое: выход за конец строки обычно не вызывает такую ошибку — Python просто берёт доступную часть.

## 5. Символ строки нельзя заменить по индексу

Строка после создания не меняется по отдельным ячейкам. Например, нельзя записать новую букву так:

```python
word = "кот"
# word[0] = "р"  # Python сообщит об ошибке
```

Квадратные скобки с индексом помогают **прочитать** символ. Они не позволяют заменить его внутри строки.

Если нужен изменённый текст, создают новую строку. Например, можно соединить части текста. Срезы, которые мы изучим ниже, как раз помогают взять нужные части строки.

## 6. Что такое срез

**Срез** позволяет взять не один символ, а несколько символов подряд.

Для простого среза пишут:

```python
строка[начало:остановка]
```

Два важных правила:

1. Индекс `начало` входит в результат.
2. Индекс `остановка` в результат **не входит**. Python остановится прямо перед ним.

Пример: `word[1:4]` берёт символы с индексами 1, 2 и 3. Символ с индексом 4 уже не попадёт в результат.

На схеме зелёным отмечены символы, которые вошли в срез. Красная граница показывает, где срез остановился.

![Срез word[1:4]: индексы 1, 2 и 3 входят, индекс 4 не входит, результат YTH](string_slice_bounds_imagegen.png)

````python
word = "PYTHON"
part = word[1:4]

print(part)
````

Разберём запись `word[1:4]`:

| Индекс | Символ | Что делает срез |
|---:|:---:|---|
| 0 | P | не берёт — мы начинаем с 1 |
| 1 | Y | берёт — это начало среза |
| 2 | T | берёт |
| 3 | H | берёт |
| 4 | O | не берёт — это граница остановки |
| 5 | N | не берёт — срез уже закончился |

Результат — новая строка `YTH`. Поэтому `print(part)` напечатает:

```text
YTH
```

## 7. Если начало или остановку не написать

В срезе начало и остановку иногда можно пропустить:

- `word[:3]` — взять с начала строки до индекса 3, не включая его;
- `word[3:]` — взять от индекса 3 до конца строки;
- `word[:]` — взять всю строку целиком.

Для слова `PYTHON` это выглядит так:

| Запись | Какие символы берём | Результат |
|---|---|---|
| `word[:3]` | индексы 0, 1, 2 | `PYT` |
| `word[3:]` | индексы 3, 4, 5 | `HON` |
| `word[:]` | все символы | `PYTHON` |

Когда начало пропущено, Python начинает с начала строки. Когда остановка пропущена, Python идёт до конца.

````python
word = "PYTHON"

print(word[:3])
print(word[3:])
print(word[:])
````

**Вывод:**

```text
PYT
HON
PYTHON
```

## 8. Отрицательные индексы в срезах

В срезах тоже можно считать с конца строки.

У `PYTHON` отрицательные индексы такие:

| Буква | P | Y | T | H | O | N |
|---|---:|---:|---:|---:|---:|---:|
| Индекс | -6 | -5 | -4 | -3 | -2 | -1 |

Например, `word[-4:-1]` начинает с `-4` и идёт до `-1`, не включая `-1`:

- `-4` — берём `T`;
- `-3` — берём `H`;
- `-2` — берём `O`;
- `-1` — останавливаемся перед `N`, поэтому `N` не берём.

Результат — `THO`.

````python
word = "PYTHON"
print(word[-4:-1])
print(word[-3:])
````

**Вывод:**

```text
THO
HON
```

Во второй строке начало равно `-3`, а остановка пропущена, поэтому берём символы от `H` до конца.

## 9. Шаг в срезе

В срезе можно указать третье число — **шаг**:

```python
строка[начало:остановка:шаг]
```

Шаг показывает, сколько индексов пропускать между выбранными символами.

Например, `word[::2]` означает: взять строку от начала до конца, выбирая каждый второй символ. В `PYTHON` будут выбраны буквы с индексами 0, 2 и 4: `P`, `T`, `O`. Получится `PTO`.

Если шаг равен 1, берётся каждый символ по порядку. Это обычное поведение среза.

Схема показывает, какие символы остаются при шаге 2 и что получается при обратном шаге `-1`.

![Срезы с шагом: word[::2] выбирает PTO, а word[::-1] переворачивает PYTHON](string_slice_steps_imagegen.png)

````python
word = "PYTHON"

print(word[::2])
print(word[1::2])
print(word[1:5:2])
````

**Вывод:**

```text
PTO
YHN
YH
```

Разбор:

- `word[::2]`: начинаем с начала, идём до конца, берём индексы 0, 2, 4 — `PTO`.
- `word[1::2]`: начинаем с индекса 1, затем берём через один — индексы 1, 3, 5 — `YHN`.
- `word[1:5:2]`: начинаем с индекса 1, останавливаемся перед 5, шагаем по 2 — индексы 1 и 3 — `YH`. Индекс 5 не включается.

## 10. Отрицательный шаг: читаем строку справа налево

Если шаг отрицательный, срез идёт справа налево.

Самая короткая запись для переворота строки:

```python
word[::-1]
```

Здесь начало и остановка пропущены. Python начинает с конца и идёт назад с шагом `-1`.

Для `PYTHON` порядок символов станет таким:

```text
N → O → H → T → Y → P
```

Результат: `NOHTYP`.

````python
word = "PYTHON"

print(word[::-1])
print(word[5:1:-1])
````

**Вывод:**

```text
NOHTYP
NOHT
```

`word[5:1:-1]` начинает с индекса 5 — буквы `N` — и движется влево. Граница 1 не включается, поэтому срез берёт индексы 5, 4, 3 и 2: `N`, `O`, `H`, `T`. Получается `NOHT`.

## 11. Пустые срезы и границы

Иногда срез не находит символов. Тогда результатом будет пустая строка `""`.

- `word[2:2]` — начало и остановка совпали, поэтому символы не берутся.
- `word[4:2]` при обычном положительном шаге — начало уже за остановкой, поэтому результат пустой.
- `word[0:100]` — граница 100 дальше конца строки. Срез спокойно возьмёт всё, что есть.

Это отличается от чтения одного символа:

- `word[100]` — ошибка, потому что индекса 100 нет;
- `word[0:100]` — ошибки нет: срез берёт доступные символы до конца.

Нулевой шаг использовать нельзя: `word[::0]` вызывает ошибку. Шаг должен быть положительным или отрицательным, но не нулём.

````python
word = "PYTHON"

print(word[2:2] == "")
print(word[4:2] == "")
print(word[0:100])
````

**Вывод:**

```text
True
True
PYTHON
```

Первые две проверки показывают, что срезы пустые. Последний срез просит символы до индекса 100, но строка заканчивается раньше. Python возвращает всю строку.

## 12. Пробелы и знаки препинания тоже имеют индексы

Индекс есть не только у букв. Пробел, запятая, точка и восклицательный знак тоже занимают место в строке.

```python
message = "Кот, привет!"
```

У этой строки:

- `К` — индекс 0;
- `о` — индекс 1;
- `т` — индекс 2;
- `,` — индекс 3;
- пробел — индекс 4;
- `п` — индекс 5.

Поэтому при работе с индексами важно считать каждый символ, включая пробелы и знаки.

````python
message = "Кот, привет!"
print(message[3])
print(message[4])
print(message[5])
````

**Вывод:**

```text
,
 
п
```

Между запятой и буквой `п` в выводе есть строка с одним пробелом. Этот пробел — отдельный символ в строке.

## Два варианта объяснения терминов

### 1. Простыми словами — чтобы понять

- **Строка** — текст в кавычках.
- **Символ** — одна буква, цифра, пробел или знак в строке.
- **Индекс** — номер места символа. Слева направо счёт начинается с 0; справа налево — с -1.
- **Длина строки** — количество символов в ней. Её показывает `len()`.
- **Срез** — способ взять часть строки.
- **Начало среза** — индекс первого символа, который нужно взять.
- **Остановка среза** — граница, перед которой нужно остановиться. Сам символ на этой границе не берётся.
- **Шаг среза** — сколько индексов пропускать между взятыми символами.
- **Отрицательный шаг** — движение по строке справа налево.
- **Неизменяемая строка** — строка, в которой нельзя заменить один символ по индексу.

### 2. С терминами — как сказать в колледже

- **Строка (`str`)** — неизменяемая упорядоченная последовательность символов Unicode.
- **Символ** — отдельная единица текста, доступная в строке по индексу.
- **Индекс** — целое число, определяющее позицию символа в последовательности. Индексация начинается с нуля; отрицательная индексация отсчитывается от конца.
- **Длина строки** — количество символов в строке, возвращаемое встроенной функцией `len()`.
- **Индексирование** — получение символа строки по его индексу с помощью квадратных скобок.
- **Срез** — операция получения части последовательности по начальной позиции, исключающей конечной позиции и шагу.
- **Начало среза** — индекс, с которого начинается выбор элементов; начальная позиция включается.
- **Остановка среза** — конечная граница выбора; элемент с этим индексом не включается.
- **Шаг среза** — расстояние между индексами выбираемых элементов.
- **Неизменяемый тип (`immutable`)** — тип данных, значение которого нельзя изменить после создания; строка относится к таким типам.
