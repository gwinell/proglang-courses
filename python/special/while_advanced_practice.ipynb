# Урок 2. Практика с циклом `while`

В этом уроке — задачи посложнее на знакомые темы: цикл `while`, счётчик, суммы, ввод данных и условия `if / elif / else`.

Новых команд в заданиях нет. 

Условия задач выдаются отдельными карточками в codeTogether.


## Самопроверка

- У каждого цикла есть понятное условие продолжения.
- Переменная, которая влияет на условие, меняется.
- Ввод считывается столько раз, сколько нужно.
- Сигнальное значение не попало в сумму или подсчёт.
- Проверки для `if / elif / else` идут в нужном порядке.
