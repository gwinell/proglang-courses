## 1. Для чего нужен `while`?

`while` — цикл. Он повторяет команды, пока условие верно.

Например, программа может снова спросить пароль, если ответ неверный. Сколько раз придётся спросить, заранее неизвестно. Для этого подходит `while`.

У цикла есть две основные части:

- **условие** — проверка, нужно ли повторять;
- **тело цикла** — команды, которые повторяются.


````python
attempt = 1
while attempt <= 3:
    print("Попытка", attempt)
    attempt = attempt + 1
print("Попытки закончились")
````

В примере `attempt <= 3` — условие. Команды с отступом образуют тело цикла.

Сначала `attempt` равно 1. После каждой попытки число увеличивается на 1. Когда оно становится 4, условие `attempt <= 3` становится ложным, и цикл заканчивается.


## 2. Как цикл проходит шаги

Python делает так:

1. Проверяет условие.
2. Если условие верно, выполняет тело цикла.
3. Возвращается к проверке условия.
4. Если условие неверно, выходит из цикла и выполняет следующие команды.

После строки `while условие:` ставится двоеточие `:`. Команды тела цикла пишутся с отступом, обычно в четыре пробела.


![Блок-схема цикла while: проверка условия, выполнение тела и возврат к проверке](while_flow.svg)

Стрелка назад показывает повторение. Цикл не переходит сразу к началу программы: он снова проверяет только условие `while`.


## 3. Условие проверяется до тела цикла

Если условие неверно уже при первой проверке, тело цикла не выполнится ни разу.


````python
number = 10
while number < 5:
    print("Эта строка не появится")
print("Программа продолжается")
````

`10 < 5` — неверно. Поэтому Python пропускает тело цикла и сразу выводит `Программа продолжается`.

Такой случай называется **ноль повторов**. Это обычная работа цикла.


## 4. Меняем значение, чтобы цикл завершился

Часто в условии используется переменная-счётчик. Чтобы цикл остановился, нужно менять эту переменную внутри цикла.

Посмотрим на значения переменной `step` по шагам.


![Пошаговая схема: значение счётчика, результат условия и действия цикла](while_trace.svg)

````python
step = 1
while step < 4:
    print("Сейчас шаг", step)
    step = step + 1
print("Готово")
````

Когда `step` становится 4, проверка `step < 4` даёт ложь. Тело больше не выполняется.

Запись `step = step + 1` значит: взять старое значение `step`, прибавить 1 и сохранить новое значение.

Короткая запись этой же команды: `step += 1`.


### Бесконечный цикл

Например, здесь `count` не меняется, поэтому `count <= 3` всё время верно:

```python
count = 1
```

## 5. Условие может проверять ответ человека

Команда `input()` получает текст. Если ответ нужно сравнивать как текст, преобразование не требуется.

В примере программа спрашивает слово, пока человек не введёт `старт`.


````python
answer = input("Напиши слово старт: ")
while answer != "старт":
    print("Пока не получилось. Попробуй ещё раз.")
    answer = input("Напиши слово старт: ")
print("Отлично, начинаем!")
````

Команда `input()` находится внутри цикла. Поэтому программа получает новый ответ после каждой ошибки.

Если оставить `input()` только перед циклом, переменная `answer` не будет меняться внутри цикла. При неверном ответе условие останется истинным, и цикл не сможет закончиться.


### `and`, `or` и `not` в условии `while`

`input()` получает текст. Если нужно проверять числа, используйте `int(input(...))`: `int()` превращает введённый текст с цифрами в целое число.

В условии можно соединять несколько проверок:

- `and` — обе части должны быть верны;
- `or` — хотя бы одна часть должна быть верна;
- `not` — меняет ответ на противоположный.

Например, число не попало в диапазон от 1 до 5, если оно меньше 1 **или** больше 5:

```python
number < 1 or number > 5
```

Пока это условие верно, программа может просить ввести число ещё раз.


````python
number = int(input("Введи число от 1 до 5: "))
while number < 1 or number > 5:
    print("Число вне диапазона. Попробуй ещё раз.")
    number = int(input("Введи число от 1 до 5: "))
print("Принято:", number)
````

## 6. `if`, `elif` и `else` внутри цикла

Внутри цикла можно поставить условие `if`. Тогда на каждом повторе программа сначала проверит условие цикла, а затем выберет действие через `if`.

Пример: вводим числа, пока не встретим 0. Для каждого ненулевого числа выводим, положительное оно или отрицательное.


![Блок-схема: цикл получает числа, а if выбирает сообщение для каждого числа](while_if_flow.svg)

````python
number = int(input("Введи число (0 — стоп): "))
while number != 0:
    if number > 0:
        print("Положительное")
    else:
        print("Отрицательное")
    number = int(input("Введи число (0 — стоп): "))
print("Готово")
````

Проверим пример по шагам. Введём `4`, затем `-2`, затем `0`:

1. Для `4` условие цикла верно. `if number > 0` верно, поэтому выводится `Положительное`.
2. Для `-2` цикл повторяется. Проверка `number > 0` неверна, поэтому выполняется `else` и выводится `Отрицательное`.
3. Для `0` условие `number != 0` неверно. Цикл заканчивается, выводится `Готово`.

**Ввод по запросам:** `4`, `-2`, `0`.

**Ожидаемый вывод:**

```text
Положительное
Отрицательное
Готово
```


### Несколько веток внутри цикла

Вместо одного `else` можно использовать `elif`. Например, проверить, равно ли число нулю, больше нуля или меньше нуля.

Важно: `if / elif / else` выбирает действие для одного повтора. После этого цикл может перейти к следующему повтору.


````python
number = -3
if number > 0:
    print("Положительное")
elif number == 0:
    print("Ноль")
else:
    print("Отрицательное")
````

В этом примере `if` находится отдельно, чтобы было легче увидеть три ветки. В цикле такое условие записывают с отступом внутри тела `while`.


## 7. Команда `break`: выйти сразу

`break` сразу завершает ближайший цикл. После него программа продолжает работу с первой команды после цикла.

`break` удобно использовать, когда особое значение означает «остановиться». Например, ноль может быть сигналом завершения ввода.


````python
total = 0
while True:
    number = int(input("Введи число (0 — стоп): "))
    if number == 0:
        break
    total = total + number
print("Сумма:", total)
````

Здесь `while True` сам по себе не заканчивается. Цикл завершится только тогда, когда число станет 0 и выполнится `break`.

Если ввести `2`, `3`, `0`, программа выведет:

```text
Сумма: 5
```


## 8. Команда `continue`: перейти к следующему повтору

`continue` пропускает оставшиеся команды текущего повтора и возвращает Python к проверке условия цикла.

В примере пропускаем числа, которые делятся на 3 без остатка. Счётчик увеличивается **до** проверки `continue`, чтобы цикл продвигался вперёд.


![Схема показывает: break выходит из цикла, continue возвращает к проверке условия](while_break_continue.svg)

````python
number = 0
while number < 8:
    number = number + 1
    if number % 3 == 0:
        continue
    print(number)
print("Готово")
````

Когда `number` становится 3 или 6, срабатывает `continue`. Команда `print(number)` для этого повтора пропускается.

Если поставить увеличение счётчика после `continue`, число 3 будет проверяться снова и снова. Поэтому сначала меняем счётчик, затем решаем, пропускать ли вывод.


## 9. Дополнительная часть: `else` у цикла `while`

У цикла тоже может быть `else`. Этот `else` связан с `while`, а не с `if`.

Он выполняется, когда условие `while` стало ложным. Если цикл остановился через `break`, этот `else` пропускается.

Эта конструкция встречается реже. Для начала достаточно понимать обычный `if / else` внутри цикла.


````python
count = 1
while count <= 3:
    print(count)
    count = count + 1
else:
    print("Счёт завершён")
````

````python
count = 1
while count <= 3:
    if count == 2:
        break
    print(count)
    count = count + 1
else:
    print("Счёт завершён")
print("Программа продолжается")
````

Во втором примере цикл завершился через `break`, поэтому `else` у цикла не выполнился. Команда после цикла всё равно выполнилась.


## 10. `while` и `for`

- Используй `for`, если заранее известно, сколько раз повторить действие или по каким значениям пройти.
- Используй `while`, если нужно повторять команды, пока условие верно, а число повторов заранее неизвестно.

Например, обратный отсчёт от числа, которое ввёл человек, удобно делать с `while`.


````python
number = 3
while number > 0:
    print(number)
    number = number - 1
print("Старт!")
````

## 11. Частые ошибки

- **Забыли изменить значение в условии.** Цикл может стать бесконечным.
- **Изменили значение в неверную сторону.** Например, число уменьшается, хотя условие ждёт его увеличения.
- **Поставили ввод перед циклом, но не внутри.** Ответ не обновляется, и условие не меняется.
- **Забыли отступ после `while`.** Команды тела цикла должны иметь одинаковый отступ.
- **Проверили условие только после тела.** В Python `while` проверяет его до каждого повтора.
- **Счётчик стоит после `continue`.** Если сработал `continue`, счётчик может не измениться и цикл застрянет.
- **Слишком рано использовали `break`.** Проверь, что он выполняется только при нужном условии.

## 12. Два варианта определений

### 1. Чтобы понять

- **Цикл `while`** — повторяет команды, пока условие верно.
- **Условие** — проверка, которая даёт ответ `True` или `False`.
- **Тело цикла** — команды с отступом, которые повторяются.
- **Повтор (итерация)** — один проход тела цикла.
- **Счётчик** — переменная, которая обычно меняется на каждом повторе.
- **Бесконечный цикл** — цикл, который не может закончиться, потому что условие всё время верно.
- **`if / elif / else` внутри цикла** — выбор действия на каждом повторе.
- **`break`** — сразу выйти из цикла.
- **`continue`** — пропустить остаток текущего повтора и начать проверку следующего.
- **Сигнальное значение** — значение, например 0, которое означает «закончить ввод».
- **`else` у цикла** — команды, которые выполняются, когда цикл закончился обычным образом, без `break`.

### 2. Чтобы сказать в колледже

- **Инструкция `while`:** составная инструкция управления потоком, которая повторно проверяет выражение и выполняет тело, пока выражение истинно.
- **Условное выражение цикла:** выражение, истинность которого проверяется перед каждым выполнением тела цикла.
- **Тело цикла (блок):** набор инструкций, принадлежащих циклу и выполняемых при истинном условии. В многострочной форме блок определяется отступом.
- **Итерация:** одно выполнение тела цикла.
- **Переменная управления циклом:** переменная, значение которой изменяется в теле или при вводе данных и влияет на достижение условия завершения.
- **Бесконечный цикл:** цикл, условие продолжения которого остаётся истинным, либо цикл без достижимого выхода.
- **Условная инструкция `if`:** выбирает одно тело из `if`, необязательных `elif` и необязательного `else` по истинности условий.
- **Инструкция `break`:** завершает ближайший охватывающий цикл и передаёт управление первой инструкции после него.
- **Инструкция `continue`:** прекращает текущую итерацию ближайшего цикла и передаёт управление к следующей проверке условия `while`.
- **Сигнальное значение:** специальное допустимое значение входных данных, которое обозначает окончание последовательности ввода и не обрабатывается как обычные данные, если условие задачи говорит об этом.
- **Клауза `else` цикла:** необязательный блок, выполняемый при обычном завершении цикла; для `while` это происходит, когда условие стало ложным. После завершения через `break` этот блок пропускается.
