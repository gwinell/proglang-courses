# Переменные и команда `print()`

В этом блокноте простыми словами разобраны переменные, команда `print()`, ввод текста и числа.

Читайте объяснение и запускайте примеры по порядку. Можно заниматься по одной теме за раз.

### Как запустить пример

Нажмите ▶ рядом с серой ячейкой или нажмите `Shift + Enter`. Результат появится под кодом.


## 1. Команда `print()`

`print()` — команда, которая выводит текст на экран.

Текст нужно написать в скобках и взять в кавычки.


![Иллюстрация: компьютер выводит результат на экран](print_screen.png)

````python
print("Привет!")
````

В примере команда `print()` выводит на экран текст `Привет!`.

Текст в программе пишется в кавычках. Команду `print` пишите маленькими буквами.


## 2. Текст и числа

Текст пишется в кавычках: `"Лена"`.

Число пишется без кавычек: `8`.

Число с дробной частью пишется с точкой: `2.5`.


````python
print("8")
print(8)
print(8 + 2)
````

На экране в обоих случаях видна цифра 8.

Но `"8"` — это текст, а `8` — число. Числа можно складывать. Текст можно соединять.


## 3. Переменные

Переменная — это имя, которому присвоено значение.

Чтобы создать переменную, напишите имя, знак `=`, затем значение:

```python
имя = значение
```

Например: `name = "Лена"`.


![Иллюстрация: значение в коробке с именной биркой](variable_box.png)

````python
name = "Лена"
print(name)
````

В первой строке переменной `name` присвоено значение `"Лена"`.

Во второй строке команда `print()` выводит значение переменной `name`.

`print(name)` выводит значение переменной. `print("name")` выводит текст `name`.


````python
name = "Лена"
print(name)
print("name")
````

### Имена переменных

Имя переменной должно подсказывать, что в ней хранится.

Примеры: `name`, `age`, `favorite_color`.

В имени не ставят пробелы. Вместо пробела можно использовать знак `_`.

`age` и `Age` — разные имена.


### Новое значение переменной

Переменной можно присвоить новое значение. После этого команда `print()` выведет новое значение.


````python
score = 3
print(score)
score = 5
print(score)
````

## 4. Аргументы

Аргумент — это значение, которое написано внутри скобок команды.

В примере `"Привет"` — аргумент команды `print()`.

Если аргументов несколько, разделите их запятыми.


````python
print("Меня зовут", "Лена")
````

Здесь у команды `print()` два аргумента: `"Меня зовут"` и `"Лена"`.

Команда выводит их через пробел.


## 5. Несколько аргументов в `print()`

Команда `print()` может выводить несколько аргументов. Разделяйте их запятыми.


````python
name = "Мира"
age = 9
print("Имя:", name)
print("Возраст:", age)
````

`name` и `age` — переменные. Команда `print()` выводит их значения вместе с текстом.

Запятые разделяют аргументы. Между ними команда ставит пробел.


### Настройка `sep`

`sep` задаёт разделитель между аргументами команды `print()`.

По умолчанию разделитель — пробел. Можно задать другой текст.


````python
print("красный", "жёлтый", "синий", sep=" | ")
print("2026", "11", "06", sep="-")
````

`sep` ставит разделитель только между аргументами.


### Настройка `end`

`end` задаёт, что команда `print()` выводит после аргументов.

По умолчанию `print()` переходит на новую строку. Это можно изменить с помощью `end`.


````python
print("Привет", end=" ")
print("мир!")
````

В первой команде `end=" "` задаёт пробел после текста. Поэтому следующая команда продолжает вывод на той же строке.

`end=""` задаёт пустой текст после вывода.


````python
print("Один", end="")
print("Два", end="!")
print("Три")
````

`sep` задаёт текст между аргументами.

`end` задаёт текст после аргументов.


````python
print("Шаг", 1, "готов", sep=" — ", end="!\n")
print("Следующая строка")
````

## 6. Знаки `\n` и `\t`

`\n` начинает новую строку внутри текста.

`\t` добавляет отступ внутри текста.

Оба знака пишутся в кавычках.


````python
print("Первая строка\nВторая строка")
print("Имя:\tМира")
````

В первом примере `\n` переносит часть текста на новую строку.


## 7. Действия с текстом

Знак `+` соединяет тексты.

Знак `*` повторяет текст указанное число раз.


````python
print("Сол" + "нце")
print("ха" * 3)
print("*" * 5)
````

Знак `+` не добавляет пробел. Чтобы получить пробел, добавьте текст `" "`.


````python
first = "Добрый"
second = "день"
print(first + " " + second)
````

## 8. Команда `input()`

`input()` показывает вопрос и ждёт, пока человек введёт ответ.

Ответ можно сохранить в переменную.


![Иллюстрация: человек вводит ответ, компьютер его получает](input_response.png)

````python
name = input("Как тебя зовут? ")
print("Привет,", name)
````

Запустите пример. Введите имя и нажмите Enter. Команда `print()` выведет приветствие.

Если ввод не работает в вашей программе, запустите следующий пример с готовым именем.


````python
name = "Саша"
print("Привет,", name)
````

### Что возвращает `input()`

Команда `input()` всегда получает ответ как текст.

Если ввести `12`, сначала это будет текст `"12"`, а не число `12`.


````python
a = "12"
b = "3"
print(a + b)
print(int(a) + int(b))
````

`"12" + "3"` соединяет тексты и выводит `123`.

`int("12") + int("3")` складывает числа и выводит `15`.


### Команда `int()`

`int()` превращает текст с целым числом в число.

Например, `int("9")` даёт число `9`.

Чтобы получить целое число от человека, используйте `int()` вместе с `input()`:

```python
age = int(input("Сколько тебе лет? "))
```


````python
age = int("9")
print(age + 1)
````

`int("9")` работает, потому что в тексте записана цифра.

`int("девять")` не работает, потому что это слово.

`float()` превращает текст с дробным числом в число. Дробная часть записывается через точку.


````python
price = float("2.5")
print(price + 1)
````

## 9. Арифметические действия

Python выполняет действия с числами.

| Знак | Действие | Пример |
|---|---|---|
| `+` | сложение | `4 + 2` даёт `6` |
| `-` | вычитание | `4 - 2` даёт `2` |
| `*` | умножение | `4 * 2` даёт `8` |
| `/` | деление | `5 / 2` даёт `2.5` |
| `//` | деление без дробной части | `5 // 2` даёт `2` |
| `%` | остаток от деления | `5 % 2` даёт `1` |
| `**` | степень | `2 ** 3` даёт `8` |


````python
a = 12
b = 5
print("Сложение:", a + b)
print("Вычитание:", a - b)
print("Умножение:", a * b)
print("Деление:", a / b)
print("Целая часть деления:", a // b)
print("Остаток:", a % b)
print("Степень:", 2 ** 3)
````

### Действия `//` и `%`

`12 // 5` показывает, сколько целых групп по 5 получится из 12. Ответ — 2.

`12 % 5` показывает остаток. Ответ — 2.


## 10. Порядок выполнения

Python выполняет команды сверху вниз.

Сначала можно присвоить значение переменной. Потом вывести его командой `print()`.


````python
city = "Казань"
print(city)
city = "Томск"
print(city)
````

Первая команда `print()` выводит значение, которое было в переменной в тот момент.

После нового присваивания переменная получает другое значение.


````python
apples = 4  # сколько яблок было
apples = apples + 2  # добавили ещё два
print(apples)  # показываем итог
````

В строке `apples = apples + 2` Python берёт старое значение `apples`, прибавляет 2 и сохраняет результат.

Текст после знака `#` — комментарий. Python его не выполняет.


````python
first_number = int(input("Введи первое целое число: "))
second_number = int(input("Введи второе целое число: "))
total = first_number + second_number
print("Сумма:", total)
````

В этом примере числа уже записаны в программе. Запустите ячейку, чтобы увидеть результат.


````python
first_number = 7
second_number = 4
total = first_number + second_number
print("Сумма:", total)
````

Программа выполняет четыре действия:

1. Сохраняет первое число.
2. Сохраняет второе число.
3. Складывает числа и сохраняет сумму.
4. Выводит сумму.


## 12. Частые ошибки

- Текст нужно писать в кавычках: `print("Привет")`.
- Проверьте, что закрыты все кавычки и скобки.
- Имя переменной нужно писать одинаково во всех строках.
- `input()` получает текст. Для вычислений преобразуйте ответ: `int(input(...))`.
- `int()` работает с цифрами, например `int("12")`. Со словами он не работает.


Для ввода числа используйте `input()` и `int()` вместе:

```python
age = int(input("Сколько тебе лет? "))
print("Через год тебе будет:", age + 1)
```

## 16. Два варианта определений

Ниже каждый термин объяснён двумя способами.

### 1. Чтобы понять

Эти объяснения короткие и простые.

- **Переменная** — имя, которому присвоено значение.
- **Значение** — данные, например текст или число.
- **Имя переменной (идентификатор)** — название, по которому программа обращается к переменной.
- **Присваивание** — запись значения в переменную с помощью `=`.
- **Тип данных** — вид значения: текст, целое число или дробное число.
- **Текст (строка)** — текст в кавычках, например `"Привет"`.
- **Функция** — готовая часть Python, которую можно вызвать.
- **Вызов функции** — команда запустить функцию. Например, `print("Привет")`.
- **Аргумент** — значение внутри скобок при вызове функции.
- **Позиционный аргумент** — аргумент, место которого важно.
- **Именованный аргумент** — аргумент с названием, например `sep="-"`.
- **`print()`** — команда, которая выводит текст на экран.
- **`sep`** — настройка, которая задаёт знак между аргументами `print()`.
- **`end`** — настройка, которая задаёт, что появится после вывода.
- **`input()`** — команда, которая получает введённый текст.
- **`int()`** — команда, которая превращает текст с целым числом в число.
- **`float()`** — команда, которая превращает текст с дробным числом в число.
- **Оператор** — знак, который задаёт действие, например `+` или `*`.
- **`//`** — деление, при котором остаётся целая часть результата.
- **`%`** — знак, который показывает остаток от деления.
- **`**`** — знак возведения в степень.
- **Комментарий** — заметка в коде после `#`. Python её пропускает.
- **`\n`** — знак перехода на новую строку внутри текста.
- **`\t`** — знак отступа внутри текста.

### 2. Чтобы сказать в колледже

Эти формулировки используют термины из программирования и Python.

- **Переменная:** имя (идентификатор), связанное в текущей области видимости с объектом Python. В Python имя указывает на объект, а не является отдельной «коробкой» с данными.
- **Объект и значение:** объект — сущность времени выполнения Python; у объекта есть тип и значение. Переменная связывает имя с объектом.
- **Идентификатор:** имя, которое соответствует правилам языка и используется для обозначения переменной, функции или другого объекта. Идентификаторы различают регистр букв.
- **Оператор присваивания:** инструкция вида `имя = выражение`; Python вычисляет выражение и связывает имя с полученным объектом. Знак `=` выполняет присваивание, а `==` сравнивает значения.
- **Тип данных:** категория объектов, которая определяет их поведение и доступные операции. Например, `str`, `int` и `float` — встроенные типы Python.
- **Строка (`str`):** неизменяемая последовательность символов Unicode. Строковый литерал обычно записывается в одинарных или двойных кавычках.
- **Функция:** вызываемый объект, который принимает аргументы, выполняет заданные действия и может вернуть результат. Встроенные функции доступны без импорта.
- **Вызов функции:** выражение с именем вызываемого объекта и скобками; аргументы указываются внутри скобок.
- **Аргумент:** значение, передаваемое функции при вызове. В определении функции соответствующее имя называется параметром.
- **Позиционный аргумент:** аргумент, который связывается с параметром по порядку в вызове.
- **Именованный аргумент:** аргумент, переданный с явным именем параметра, например `sep="-"`.
- **Встроенная функция `print()`:** записывает текстовые представления переданных объектов в текстовый поток. По умолчанию разделяет их строкой `sep=" "`, завершает строкой `end="\n"` и пишет в стандартный поток вывода. Функция возвращает `None`.
- **Параметр `sep`:** именованный параметр `print()`, задающий строку-разделитель между объектами. Значение по умолчанию — один пробел.
- **Параметр `end`:** именованный параметр `print()`, задающий строку после последнего объекта. Значение по умолчанию — символ новой строки `"\n"`.
- **Встроенная функция `input()`:** при наличии выводит приглашение, читает одну строку из стандартного потока ввода, удаляет завершающий символ новой строки и возвращает результат типа `str`.
- **Тип `int` и функция-конструктор `int()`:** представляют целые числа произвольной точности и позволяют создавать целое число из совместимого значения, например из текста с цифрами.
- **Тип `float` и функция-конструктор `float()`:** представляют числа с плавающей точкой; значения хранятся в двоичном формате и могут быть приближёнными.
- **Арифметический оператор:** знак, который задаёт операцию над операндами. Для чисел `+`, `-`, `*`, `/`, `//`, `%` и `**` означают сложение, вычитание, умножение, деление, деление с округлением вниз, остаток и возведение в степень.
- **Конкатенация строк:** операция соединения строк. Для строк её выполняет оператор `+`.
- **Повторение последовательности:** операция создания повторённой последовательности. Например, строка умножается на целое число: `"*" * 3`.
- **Деление `//`:** операция деления с округлением результата вниз до ближайшего целого значения.
- **Остаток `%`:** операция получения остатка деления. Для чисел Python связывает результат с делением равенством `a == (a // b) * b + (a % b)`.
- **Возведение в степень `**`:** бинарная операция возведения левого операнда в степень правого.
- **Управляющая последовательность:** запись обратной косой черты с символом внутри строкового литерала, обозначающая специальный символ. Например, `\n` — новая строка, `\t` — горизонтальная табуляция.
- **Комментарий:** текст от символа `#` до конца физической строки вне строкового литерала; интерпретатор не выполняет этот текст как код.
