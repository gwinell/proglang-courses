# Часть 1. Цикл работает со списком

## 1. Что такое список

**Простыми словами:** список — это несколько значений, записанных по порядку. Представь полку с пронумерованными ячейками. В каждой ячейке лежит одно значение.

В Python список записывают в квадратных скобках. Элементы разделяют запятыми:

````python
animals = ["кот", "пёс", "ёж", "лиса"]
print(animals)
````

На экране будет:

```text
['кот', 'пёс', 'ёж', 'лиса']
```

У каждой ячейки есть **индекс** — её номер. В Python индексы начинаются с нуля:

| Индекс | Элемент |
|---:|---|
| `0` | `"кот"` |
| `1` | `"пёс"` |
| `2` | `"ёж"` |
| `3` | `"лиса"` |

Список хранит элементы по порядку. Цикл может брать эти элементы один за другим.

## 2. Перебираем сами элементы

**Простыми словами:** цикл `for` берёт первый элемент списка, выполняет команды с ним, затем берёт следующий. Так продолжается, пока элементы не закончатся.

```python
for временное_имя in список:
    команды
```

Временное имя — это переменная. На каждом круге в неё попадает очередной элемент.

### Схема обхода списка

Зелёная рамка показывает, какой элемент цикла обрабатывает сейчас. Индексы под ячейками помогают увидеть порядок.

![Схема обхода списка циклом for: текущая ячейка перемещается слева направо, индексы идут от 0 до 3](lists_loop_traversal_imagegen.png)

````python
animals = ["кот", "пёс", "ёж", "лиса"]

for animal in animals:
    print(animal)
````

**Что сделает программа:**

1. Запишет `"кот"` в переменную `animal` и напечатает его.
2. Запишет `"пёс"` в `animal` и напечатает его.
3. Запишет `"ёж"` в `animal` и напечатает его.
4. Запишет `"лиса"` в `animal` и напечатает её.
5. Когда элементы закончатся, цикл завершится.

Вывод будет таким:

```text
кот
пёс
ёж
лиса
```

Переменная `animal` — не весь список. В ней находится только один текущий элемент.

### Переменная цикла меняется на каждом круге

В этом примере мы напечатаем и номер круга, и текущее животное. Переменную `step` создаём отдельно: Python сам её не добавляет.

````python
animals = ["кот", "пёс", "ёж"]
step = 1

for animal in animals:
    print("Круг", step, "— сейчас обрабатываем:", animal)
    step = step + 1
````

После первого круга `step` станет равна 2. После второго — 3. Это обычная переменная: мы меняем её командой `step = step + 1`.

Вывод:

```text
Круг 1 — сейчас обрабатываем: кот
Круг 2 — сейчас обрабатываем: пёс
Круг 3 — сейчас обрабатываем: ёж
```

## 3. Условие внутри цикла

**Простыми словами:** на каждом круге можно задать вопрос про текущий элемент. Команда `if` проверит этот вопрос. Если ответ «да», программа выполнит команды внутри `if`.

Например, оставим на экране только числа больше пяти:

````python
numbers = [4, 7, 2, 9, 6]

for number in numbers:
    if number > 5:
        print(number, "больше 5")
````

Цикл проверит числа по порядку:

| Текущее число | Проверка `number > 5` | Что делает программа |
|---:|---|---|
| 4 | Нет | Ничего не печатает |
| 7 | Да | Печатает `7 больше 5` |
| 2 | Нет | Ничего не печатает |
| 9 | Да | Печатает `9 больше 5` |
| 6 | Да | Печатает `6 больше 5` |

Условие относится к одному элементу на текущем круге. На следующем круге в переменной `number` уже будет другое число.

### `if` и `else` внутри цикла

`else` задаёт действия на случай, когда условие не выполнилось. Эти ветки проверяются заново для каждого элемента.

````python
temperatures = [18, 24, 15]

for temperature in temperatures:
    if temperature >= 20:
        print(temperature, "— тепло")
    else:
        print(temperature, "— прохладно")
````

Вывод:

```text
18 — прохладно
24 — тепло
15 — прохладно
```

Цикл не выбирает одну ветку раз и навсегда. Он каждый раз проверяет `if` для текущего значения.

## 4. Считаем элементы, которые подходят

**Простыми словами:** счётчик — это переменная, которая хранит количество найденных элементов. Сначала найденных элементов ноль. Когда встречаем подходящий элемент, увеличиваем счётчик на один.

Посчитаем, сколько раз в списке встречается оценка 5:

````python
grades = [5, 3, 5, 4, 2, 5]
count = 0

for grade in grades:
    if grade == 5:
        count = count + 1

print("Пятёрок:", count)
````

Разбор:

1. `count = 0` — пока пятёрок не найдено.
2. Цикл берёт очередную оценку.
3. Если оценка равна 5, программа прибавляет 1 к `count`.
4. После окончания цикла печатается итог.

В списке три пятёрки, поэтому программа напечатает:

```text
Пятёрок: 3
```

Важно: итог печатается **после цикла**, поэтому команда `print()` не повторяется для каждой оценки.

## 5. Находим элементы и сохраняем их

Иногда мало просто напечатать найденное значение. Его нужно собрать в отдельном списке.

**Простыми словами:** сначала создаём пустой список `big_numbers`. Затем проверяем числа по одному. Если число больше пяти, добавляем его в новый список командой `.append()`.

`.append(значение)` добавляет значение в конец списка.

````python
numbers = [4, 7, 2, 9, 6]
big_numbers = []

for number in numbers:
    if number > 5:
        big_numbers.append(number)

print(big_numbers)
````

Новый список будет таким:

```text
[7, 9, 6]
```

Старый список `numbers` содержит `[4, 7, 2, 9, 6]`. Новый список `big_numbers` содержит только числа, которые прошли проверку. Мы не удаляли числа из старого списка.

### Что происходит с новым списком на каждом круге

Начало: `big_numbers = []`.

| Текущий элемент | Он больше 5? | Новый список после этого круга |
|---:|---|---|
| 4 | Нет | `[]` |
| 7 | Да | `[7]` |
| 2 | Нет | `[7]` |
| 9 | Да | `[7, 9]` |
| 6 | Да | `[7, 9, 6]` |

`append()` срабатывает только там, где условие истинно.

## 6. Элемент и индекс — это разные вещи

В примере `for animal in animals` переменная `animal` получает **сам элемент**: например, строку `"пёс"`.

Иногда нужны номера ячеек. Тогда можно пройти по числам от нуля до последнего индекса с помощью `range(len(animals))`:

- `len(animals)` сообщает, сколько элементов в списке;
- `range(...)` даёт номера по порядку;
- `animals[index]` берёт элемент по номеру.

````python
animals = ["кот", "пёс", "ёж"]

for index in range(len(animals)):
    print("Индекс:", index, "Элемент:", animals[index])
````

`len(animals)` равно 3. Поэтому `range(len(animals))` выдаёт индексы `0`, `1`, `2`.

Вывод:

```text
Индекс: 0 Элемент: кот
Индекс: 1 Элемент: пёс
Индекс: 2 Элемент: ёж
```

Используй обычный перебор `for item in items`, когда нужен сам элемент. Используй индексы, когда нужен номер позиции или нужно заменить значение в конкретной ячейке.

## 7. Меняем элемент списка по индексу

**Простыми словами:** список можно менять. Если знаем номер ячейки, можем положить в неё другое значение.

Запись `animals[index] = новое_значение` заменяет элемент с этим индексом.

````python
animals = ["кот", "пёс", "ёж"]

for index in range(len(animals)):
    animals[index] = animals[index] + "!"

print(animals)
````

Цикл проходит индексы 0, 1 и 2. На каждом шаге он берёт значение из ячейки и записывает туда новое значение с восклицательным знаком.

Результат:

```text
['кот!', 'пёс!', 'ёж!']
```

Здесь мы меняем исходный список. Если исходный список нужно сохранить, создай новый список и добавляй результаты в него.

### Почему опасно менять длину списка во время обхода

Пока цикл перебирает список, не удаляй из него элементы прямо в том же цикле. После удаления элементы сдвигаются, и цикл может пропустить значение.

Для отбора обычно безопаснее создать новый список через `append()`:

```python
source = [2, 5, 8, 3]
result = []

for number in source:
    if number % 2 == 0:
        result.append(number)
```

Здесь `source` остаётся прежним, а подходящие элементы записываются в `result`.

# Часть 2. Создаём список по списку

## 8. Сначала понятная длинная запись

Задача: получить список, где каждое число удвоено.

Сначала создаём пустой список. Затем цикл берёт число, вычисляет результат и добавляет его в новый список.

````python
numbers = [2, 4, 6]
doubled = []

for number in numbers:
    new_number = number * 2
    doubled.append(new_number)

print(doubled)
````

После каждого круга:

| `number` | `new_number` | `doubled` после `append()` |
|---:|---:|---|
| 2 | 4 | `[4]` |
| 4 | 8 | `[4, 8]` |
| 6 | 12 | `[4, 8, 12]` |

Вывод программы: `[4, 8, 12]`. В этой длинной записи хорошо видно все шаги.

## 9. Генератор списка: короткая запись

**Простыми словами:** генератор списка — это короткая запись, которая создаёт новый список. Он берёт элементы старого списка, выполняет действие и складывает результаты в новый.

В английских материалах это называют **list comprehension**. Несмотря на русское слово «генератор», такая запись сразу создаёт готовый список.

Общая форма:

```python
[выражение for элемент in список]
```

В квадратных скобках записано выражение, которое вычисляется для каждого элемента.

````python
numbers = [2, 4, 6]
doubled = [number * 2 for number in numbers]
print(doubled)
````

Эта строка:

```python
doubled = [number * 2 for number in numbers]
```

делает то же самое, что длинная запись выше:

1. берёт `2`, вычисляет `2 * 2`, записывает `4`;
2. берёт `4`, вычисляет `4 * 2`, записывает `8`;
3. берёт `6`, вычисляет `6 * 2`, записывает `12`.

Результат — готовый список `[4, 8, 12]`. Сначала учись читать длинную запись, затем сокращай её.

### Схема: цикл и условие собирают новый список

Числа из исходного списка проверяются по одному. В новый список попадают только чётные числа.

![Схема отбора: числа 2, 5, 8, 3 проверяются на чётность, в новый список попадают 2 и 8](lists_filter_pipeline_imagegen.png)

## 10. Добавляем условие отбора

Если нужно взять не все элементы, после части `for ... in ...` можно добавить `if`.

**Простыми словами:** сначала программа проверяет условие. Если оно истинно, элемент попадает в новый список. Если ложно — элемент пропускается.

Форма записи:

```python
[элемент for элемент in список if условие]
```

````python
numbers = [2, 5, 8, 3]
even_numbers = [number for number in numbers if number % 2 == 0]
print(even_numbers)
````

Проверим каждое число:

- `2 % 2 == 0` — да, число чётное, добавить 2;
- `5 % 2 == 0` — нет, пропустить 5;
- `8 % 2 == 0` — да, добавить 8;
- `3 % 2 == 0` — нет, пропустить 3.

Программа напечатает `[2, 8]`.

Запомни порядок: **что положить** → `for` **откуда брать** → `if` **какие элементы оставить**.

### Условие-отбор и выбор нового значения — не одно и то же

В этой записи `if` стоит в конце. Он решает, попадёт ли элемент в новый список:

```python
[number for number in numbers if number > 0]
```

В другом варианте `if` выбирает, какое значение записать. Такой вариант длиннее, и сейчас его достаточно просто узнавать:

```python
[number if number > 0 else 0 for number in numbers]
```

В первой записи число либо попадает в список, либо нет. Во второй записи для каждого элемента создаётся результат: положительное число или ноль.

# Часть 3. Генераторы значений

## 11. Важно различать два похожих названия

Есть:

1. **генератор списка** — коротко создаёт готовый список в квадратных скобках: `[x * 2 for x in numbers]`;
2. **генератор значений** — выдаёт значения постепенно, когда программа их запрашивает.

Квадратные скобки создают список сразу. Круглые скобки в похожей записи создают генератор значений:

```python
[x * 2 for x in numbers]  # готовый список
(x * 2 for x in numbers)  # генератор значений
```

Это разные объекты, хотя записи похожи.

## 12. Генератор значений: выдаёт по одному

**Простыми словами:** генератор похож на автомат. Ты просишь следующее значение — автомат вычисляет его и отдаёт. Пока его не попросили, следующее значение не вычисляется.

У генератора есть текущая позиция. После выдачи значения он запоминает, где остановился. При следующей просьбе продолжает со следующего элемента.

![Генератор выдаёт числа 10, 20 и 30 по запросу, останавливаясь между выдачами](generator_yield_pause_imagegen.png)

### Генераторное выражение в действии

Сравним квадратные и круглые скобки. Здесь переменная `numbers` — обычный список:

````python
numbers = [2, 4, 6]

ready_list = [number * 2 for number in numbers]
values_generator = (number * 2 for number in numbers)

print(ready_list)
print(values_generator)
````

Первая команда `print()` покажет готовый список:

```text
[4, 8, 12]
```

Вторая покажет описание объекта генератора, например `<generator object ...>`. Это не сами значения. Генератор надо перебирать или просить у него значения.

### Запрашиваем значения через `next()`

Команда `next(generator)` просит у генератора одно следующее значение. Каждый вызов двигает генератор вперёд на один элемент.

````python
numbers = [2, 4, 6]
values_generator = (number * 2 for number in numbers)

print(next(values_generator))
print(next(values_generator))
print(next(values_generator))
````

Результат:

```text
4
8
12
```

Первый вызов получил 4. Второй — 8. Третий — 12. Генератор не возвращается к началу автоматически.

### Перебираем генератор циклом `for`

Обычно вручную писать много вызовов `next()` не нужно. Цикл `for` сам просит у генератора очередное значение, выполняет тело и просит следующее.

````python
numbers = [2, 4, 6]
values_generator = (number * 2 for number in numbers)

for value in values_generator:
    print(value)
````

Цикл напечатает `4`, `8`, `12`, а затем завершится, когда генератору больше нечего выдавать.

## 13. Генератор можно исчерпать

**Простыми словами:** когда все значения уже выданы, генератор заканчивается. Повторный обход того же объекта ничего не напечатает.

Чтобы получить значения заново, создай новый генератор.

````python
numbers = [1, 2, 3]
values_generator = (number * 10 for number in numbers)

print(list(values_generator))
print(list(values_generator))

new_generator = (number * 10 for number in numbers)
print(list(new_generator))
````

Вывод:

```text
[10, 20, 30]
[]
[10, 20, 30]
```

Первый `list()` забрал все значения. Поэтому второй раз старый генератор уже пуст. `new_generator` — новый объект, он начинает сначала.

### Список и генератор — сравнение

| Вопрос | Список | Генератор значений |
|---|---|---|
| Когда появляются результаты? | Сразу создаётся весь список | Значения выдаются по запросу |
| Как записать похожую обработку? | `[x * 2 for x in numbers]` | `(x * 2 for x in numbers)` |
| Можно обратиться к элементу по индексу? | Да, например `items[0]` | Обычно нет; значения получают по порядку |
| Можно пройти сначала снова? | Да, по списку можно запускать новый цикл | Сам генератор обычно одноразовый |
| Когда удобно? | Когда нужны все результаты и повторный доступ | Когда удобно обрабатывать значения постепенно |

Генератор не «лучше» списка всегда. Выбирай по задаче.

## 14. Генераторная функция и команда `yield`

Генератор можно написать как функцию. Для этого внутри функции используют `yield`.

**Простыми словами:** `return` завершает функцию и отдаёт результат. `yield` отдаёт одно значение, но оставляет функцию на паузе. Когда программа запросит следующее значение, функция продолжит работу с места остановки.

````python
def count_to_three():
    yield 1
    yield 2
    yield 3

for number in count_to_three():
    print(number)
````

Вывод:

```text
1
2
3
```

Порядок работы:

1. Цикл начинает работу функции `count_to_three()`.
2. Функция доходит до `yield 1` и отдаёт число 1. На этом месте она ставит выполнение на паузу.
3. Цикл просит следующее число. Функция продолжает с места паузы и доходит до `yield 2`.
4. Затем так же выдаётся 3.
5. После последнего `yield` функция заканчивается, цикл завершается.

### Значения можно вычислять во время работы функции

В генераторе не обязательно заранее записывать все результаты. Функция может вычислить очередное значение, выдать его и перейти на паузу.

````python
def doubled_values(numbers):
    for number in numbers:
        result = number * 2
        yield result

scores = [3, 5, 2]

for value in doubled_values(scores):
    print(value)
````

На каждом шаге функция берёт один балл, умножает его на два и выдаёт результат через `yield`.

Вывод:

```text
6
10
4
```

`yield` не печатает значение на экран. Он передаёт это значение тому, кто перебирает генератор. В нашем примере значение получает цикл `for`, а печатает уже команда `print(value)`.

### Генераторная функция начинается при запросе значения

Вызов генераторной функции создаёт генератор. Команды в её теле начинают выполняться, когда мы запрашиваем значение через `next()` или запускаем цикл.

````python
def small_generator():
    print("Начало функции")
    yield "первое значение"
    print("Продолжение функции")
    yield "второе значение"

stream = small_generator()
print("Генератор создан")
print(next(stream))
print(next(stream))
````

Примерный вывод:

```text
Генератор создан
Начало функции
первое значение
Продолжение функции
второе значение
```

Сначала создан объект генератора. Затем первый `next()` запускает тело и выполняет его до первого `yield`. Следующий `next()` продолжает выполнение до второго `yield`.

## 15. Зачем нужен генератор

Генератор полезен, когда значения удобно получать постепенно:

- не нужно заранее собирать весь результат в список;
- следующий результат вычисляется тогда, когда он понадобился;
- можно обработать поток значений по одному.

Для маленького списка с тремя числами обычный список часто понятнее. Генератор не требуется добавлять в программу «просто потому, что он есть».

Если задача требует обращаться к результатам по индексам, повторно читать их или показать весь список, список обычно удобнее.

## 16. Частые ошибки и как их понять

### Ошибка 1. Ждать, что генератор сразу покажет список

```python
values = (number * 2 for number in [1, 2, 3])
print(values)
```

`print(values)` показывает объект генератора, а не его значения. Чтобы прочитать значения, используй цикл или `list(values)`.

### Ошибка 2. Пытаться получить генератор по индексу

```python
values = (number * 2 for number in [1, 2, 3])
# values[0]  # так к генератору обычно не обращаются
```

Генератор выдаёт значения по порядку. Если нужен доступ по индексу, создай список: `ready = list(values)`.

### Ошибка 3. Повторно обходить уже пустой генератор

Если генератор уже прошёл все значения, новый цикл не начнёт его с самого начала. Создай генератор заново.

### Ошибка 4. Перепутать генератор списка с генератором значений

- квадратные скобки `[...]` — готовый список;
- круглые скобки `(...)` с `for` внутри — генератор значений.

# Короткое повторение простыми словами

- `for animal in animals` — взять каждый элемент списка по очереди.
- `if` внутри цикла — проверить текущий элемент.
- `count = count + 1` — увеличить счётчик, если нашли нужное.
- `result.append(value)` — добавить значение в конец списка.
- `range(len(items))` — пройти по индексам списка.
- `items[index] = value` — заменить элемент по индексу.
- `[выражение for item in items]` — сразу создать новый список.
- `(выражение for item in items)` — создать генератор, который выдаёт значения по запросу.
- `yield value` — выдать одно значение из генераторной функции и остановиться до следующего запроса.

## Список и обход цикла

**Список (`list`)** — изменяемая упорядоченная коллекция объектов Python. Доступ к элементам списка возможен по целочисленному индексу.

**Итерация** — последовательное получение элементов и выполнение действий для каждого из них.

**Цикл `for`** выполняет тело для элементов итерируемого объекта. При записи `for item in items` переменная `item` связывается с очередным значением.

**Индекс** — целочисленная позиция элемента в индексируемой последовательности. В списках Python индексация начинается с 0.

**Итерируемый объект (iterable)** — объект, который предоставляет последовательность элементов для перебора. Список является итерируемым объектом.

**Условный отбор** — проверка предиката для каждого элемента с обработкой только элементов, для которых условие истинно.

**Аккумулятор** — переменная или коллекция, в которой накапливается промежуточный или итоговый результат вычисления. Счётчик — частный случай аккумулятора.

## Создание списков

**`append()`** — метод списка, добавляющий один объект в конец списка.

**Преобразование элементов (mapping)** — создание результата для каждого элемента входной последовательности. Например, умножение каждого числа на 2.

**Фильтрация (filtering)** — отбор тех элементов входной последовательности, которые удовлетворяют условию.

**Включение списка (list comprehension)** — синтаксическая конструкция Python для создания списка из выражения, цикла и необязательного условия. Пример:

```python
[number * 2 for number in numbers]
```

Результат list comprehension — новый список, вычисляемый при выполнении этой конструкции.

## Итераторы и генераторы

**Итератор (iterator)** — объект, который представляет последовательность значений и хранит своё текущее положение. Следующее значение получают функцией `next()`.

**Генератор (generator)** — специальный итератор, который производит значения по мере запроса. После выдачи очередного значения он сохраняет состояние выполнения.

**Генераторное выражение (generator expression)** — выражение вида `(выражение for элемент in источник)`, создающее генератор.

**Генераторная функция (generator function)** — функция, в теле которой есть `yield`. Вызов такой функции создаёт генераторный объект; тело функции выполняется по мере получения значений.

**`yield`** — оператор, который приостанавливает генераторную функцию и передаёт очередное значение вызывающему коду. При следующем запросе выполнение продолжается после `yield`.

**Ленивые вычисления (lazy evaluation)** — вычисление значения не заранее, а в момент, когда это значение требуется.

**Однократный проход** — свойство обычного итератора или генератора: после исчерпания его нельзя автоматически запустить с начала. Для повторного прохода требуется получить новый итератор или создать новый генератор.
