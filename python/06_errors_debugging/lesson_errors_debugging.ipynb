## 1. Ошибка — это подсказка

Ошибки можно условно разделить на три группы:

- **Синтаксическая ошибка**: Python не может прочитать программу, например, пропущено двоеточие после `if`.
- **Ошибка во время выполнения**: программа запустилась, но в процессе возникла проблема, например, деление на ноль.
- **Логическая ошибка**: программа выполняется, но считает или делает не то, что задумано.

Синтаксическую ошибку нужно исправить в коде до запуска. Ошибки времени выполнения иногда можно обработать. Логическую ошибку обычно находят, проверяя шаги и значения.


### Синтаксическая ошибка

Попробуй убрать `#` и запустить следующую строку. Python сообщит, где встретил проблему:


```python
# if 3 > 1
#     print("Три больше одного")
```

Здесь после условия `if` пропущено двоеточие. Пока строка закомментирована, ячейка выполняется без ошибки.

### Ошибка во время выполнения

Следующий пример синтаксически правильный, но делить на ноль нельзя. Удали `#`, чтобы увидеть сообщение Python.


```python
# result = 10 / 0
# print(result)
```

### Логическая ошибка

Код может выполняться без исключения, но давать неверный ответ. Например, площадь прямоугольника нужно считать умножением сторон, а не сложением.


```python
width = 5
height = 3

wrong_area = width + height
right_area = width * height

print("Ошибка в формуле:", wrong_area)
print("Правильная площадь:", right_area)
```

## 2. Читаем traceback

Когда во время выполнения возникает необработанная ошибка, Python показывает **traceback** — путь вызовов, который привёл к ней.

Упрощённо читай traceback снизу вверх:

1. Последняя строка называет тип ошибки и объясняет, что произошло.
2. Строка перед ней указывает файл и номер строки, где возникла ошибка.
3. Строки выше показывают, через какие вызовы программа дошла до этого места.

Например, в сообщении `ZeroDivisionError: division by zero` понятно, что программа делила на ноль. Сначала найди последнюю строку своего кода в traceback, а затем проверь значения, участвовавшие в операции.


### Посмотрим на настоящий traceback

Удали `#` и выполни ячейку. После появления ошибки найди её тип и строку кода, на которой она возникла.


```python
# price = 100
# people = 0
# print(price / people)
```

В traceback обычно есть и строки внутреннего кода Python. Для начала сосредоточься на последней строке с типом ошибки и на ближайшей строке своего файла или блокнота.

## 3. Обрабатываем ожидаемые ошибки: `try` и `except`

Код, в котором может возникнуть ожидаемая ошибка, помести в блок `try`. Если возникнет указанное исключение, выполнится блок `except`.


```python
try:
    number = int(input("Введи целое число: "))
except ValueError:
    print("Это не похоже на целое число.")
```

`ValueError` возникает, когда значение имеет подходящий тип, но его содержимое нельзя преобразовать. Например, `int("пять")` не сработает.

Перехватывай конкретный тип ошибки, который ожидаешь. Если написать только `except:`, можно случайно спрятать совсем другую проблему в программе.


### Ошибка деления на ноль

Можно обработать конкретную ошибку и объяснить пользователю, как исправить ввод.


```python
try:
    numerator = int(input("Делимое: "))
    denominator = int(input("Делитель: "))
    result = numerator / denominator
except ValueError:
    print("Нужно вводить целые числа.")
except ZeroDivisionError:
    print("На ноль делить нельзя.")
else:
    print("Результат:", result)
```

В `except` можно сохранить объект ошибки через `as`. Его сообщение удобно показать или использовать при отладке.


```python
try:
    age = int("много")
except ValueError as error:
    print("Подробность:", error)
```

Если разные ошибки требуют одинаковой реакции, их можно перечислить в кортеже. Но не объединяй исключения, если для них нужны разные подсказки.


```python
values = ["15", "пять", None]

for value in values:
    try:
        number = int(value)
    except (ValueError, TypeError):
        print("Не получилось прочитать число:", value)
    else:
        print("Ты ввёл:", number)
```

## 4. `else` и `finally`

У конструкции есть дополнительные блоки:

- `else` запускается, только если в `try` не возникло ошибки;
- `finally` запускается в любом случае: была ошибка или нет.

`else` удобно использовать для действий, которые должны происходить только после успешного выполнения `try`.


![Пути выполнения блоков try, except, else и finally](assets/try_except_flow.png)


```python
try:
    number = int(input("Введи число: "))
except ValueError:
    print("Нужно ввести число.")
else:
    print("Квадрат числа:", number ** 2)
finally:
    print("Проверка завершена.")
```

Попробуй ввести сначала число, а затем слово. Обрати внимание: сообщение из `finally` появится в обоих случаях.

Не помещай в `finally` обычные вычисления результата. Этот блок обычно нужен для завершения или уборки ресурсов; к файлам мы вернёмся в отдельном уроке.


## 5. Проверяем данные и поднимаем ошибку

Иногда функция должна остановиться, если ей передали недопустимое значение. Команда `raise` поднимает исключение. Для неподходящего значения часто используют `ValueError`.


```python
def calculate_discount(price, percent):
    if price < 0:
        raise ValueError("Цена не может быть отрицательной")
    if not 0 <= percent <= 100:
        raise ValueError("Скидка должна быть от 0 до 100 процентов")
    return price * (1 - percent / 100)

print(calculate_discount(1000, 15))
```

Проверяй условия, которые важны для функции. Ошибку можно перехватить там, где программа знает, как объяснить её пользователю.


```python
try:
    price = float(input("Цена товара: "))
    discount = float(input("Скидка в процентах: "))
    total = calculate_discount(price, discount)
except ValueError as error:
    print("Не удалось посчитать:", error)
else:
    print(f"Цена со скидкой: {total:.2f}")
```

**Не используй исключения вместо обычного условия.** Если легко проверить `denominator != 0`, это можно сделать до деления. Обработка исключений нужна для ситуаций, когда операция может не получиться, несмотря на разумные проверки.

## 6. Ищем причину сбоя

Когда программа ведёт себя не так, действуй по шагам:

1. **Повтори проблему.** Запиши, какие данные к ней приводят.
2. **Найди место сбоя.** Посмотри на тип ошибки и строку в traceback.
3. **Проверь входные данные.** Выведи значения и их типы через `print()` и `type()`.
4. **Сократи пример.** Оставь только несколько строк, которые воспроизводят проблему.
5. **Исправь причину и повтори тот же пример.**

Например, если список неожиданно пустой, проверь, что находится в нём до и после цикла.


```python
scores = [5, 4, 3]
total = 0

for score in scores:
    print("Сейчас score =", score, "тип:", type(score))
    total += score

print("Сумма:", total)
```

Отладочный вывод помогает увидеть значения по ходу программы. Убери временные `print()` после того, как нашёл проблему, или оставь только полезные сообщения.

### `assert` для проверки предположений

`assert` проверяет условие и сообщает об ошибке, если оно ложно. Используй его во время разработки, чтобы проверить собственные предположения. Для проверки пользовательского ввода применяй обычное условие и подходящее сообщение.


```python
def square(number):
    return number * number

assert square(3) == 9
assert square(0) == 0
print("Проверки пройдены")
```

Если проверка не прошла, Python покажет `AssertionError`. Важно: `assert` не заменяет проверку данных, которые ввёл пользователь.

## 7. Частые типы ошибок

| Ошибка | Частая причина | Что проверить |
|---|---|---|
| `NameError` | Переменная не создана или имя написано иначе | Имя и место создания переменной |
| `TypeError` | Операция не подходит для такого типа данных | Типы значений через `type()` |
| `ValueError` | Преобразование не удалось | Содержимое строки перед `int()` или `float()` |
| `IndexError` | Индекс выходит за границы списка | Длину списка и значение индекса |
| `KeyError` | Такого ключа нет в словаре | Какие ключи есть; можно ли использовать `get()` |
| `ZeroDivisionError` | Деление на ноль | Значение делителя |
| `IndentationError` | Неверный отступ | Отступы строк внутри блока |
| `SyntaxError` | Нарушено правило записи Python | Строку и символ, указанные в сообщении |

Одинаковый тип ошибки может возникнуть по разным причинам. Используй traceback и проверяй реальные значения.
