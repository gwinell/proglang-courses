## 1. Зачем нужен цикл

Если нужно вывести приветствие один раз, достаточно одной команды. А если сто раз? Копировать команду неудобно: легко ошибиться, и программу трудно менять. Цикл позволяет написать повторяемое действие один раз.

Посмотри сначала на обычные повторяющиеся команды:

````python
print("Привет!")
print("Привет!")
print("Привет!")
````

Это сработает, но при большом количестве повторов запись станет длинной. Цикл `for` выполняет команды внутри себя несколько раз — по одному разу для каждого очередного значения.

### Первый цикл

Прочитай строку буквально: «для каждого числа `i` из последовательности `range(3)` выполни команду с отступом».

````python
for i in range(3):
    print("Привет!")
````

**Разбор частей:**

- `for` — начинает цикл;
- `i` — переменная цикла, в которую по очереди попадает значение;
- `in` — слово «из»;
- `range(3)` — последовательность из трёх чисел: `0`, `1`, `2`;
- `:` завершает строку-заголовок;
- команда с отступом — тело цикла, оно повторяется.

Отступ — четыре пробела — показывает Python, какие команды относятся к циклу. В Jupyter после двоеточия нажми Enter: отступ обычно добавится сам.

### Мини-эксперимент

Измени `range(3)` на `range(1)`, запусти ячейку. Сколько раз появилось приветствие? А что будет при `range(0)`?

````python
for i in range(3):
    print("Шаг цикла")
````

## 2. Что происходит на каждой итерации

Один полный проход тела цикла называется **итерацией**. Перед каждой итерацией Python получает следующее значение и записывает его в переменную цикла. Затем выполняет тело. После этого переходит к следующему значению.

![Схема шагов цикла for](for_steps.png)

Посмотрим не только на сообщение, но и на значение переменной:

````python
for i in range(1, 5):
    print("Сейчас i равно", i)
````

Тут четыре итерации. В первой `i = 1`, во второй `i = 2`, затем `3` и `4`. После этого значения заканчиваются, и цикл завершается.

| Номер итерации | Значение `i` | Что печатает команда |
|---:|---:|---|
| 1 | 1 | Сейчас i равно 1 |
| 2 | 2 | Сейчас i равно 2 |
| 3 | 3 | Сейчас i равно 3 |
| 4 | 4 | Сейчас i равно 4 |

Важно: `i` — обычная переменная. Имя не особенное: можно написать `number`, `step` или `letter`. Выбирай название, которое объясняет, что хранится. После завершения цикла в переменной останется последнее присвоенное значение, если цикл выполнился хотя бы раз. Обычно внутри цикла мы используем её для текущего шага.

### Прочитай цикл по строкам

1. Возьми первое значение `1`, положи его в `number`.
2. Напечатай `number` и его квадрат.
3. Возьми следующее значение `2` и снова выполни тело.
4. Продолжай, пока значения не закончатся.

````python
for number in range(1, 4):
    print("Число:", number)
    print("Квадрат:", number * number)
````

### Задание на предсказание

Не запускай сразу: выпиши, что напечатает каждая строка и сколько всего будет строк. Потом запусти и сравни.

````python
for step in range(2):
    print("Начало", step)
    print("Продолжаем")
print("Цикл закончился")
````

## 3. Как устроен `range()`

`range()` задаёт целые числа, которые цикл будет брать по очереди. Есть три удобные формы.

### Форма 1: `range(stop)`

Начинает с нуля и останавливается **перед** `stop`:

- `range(4)` даёт `0, 1, 2, 3`;
- это четыре значения, потому что ноль тоже считается.

### Форма 2: `range(start, stop)`

Начинает с `start`, останавливается перед `stop`:

- `range(2, 6)` даёт `2, 3, 4, 5`;
- первое значение включается, граница `stop` не включается.

### Форма 3: `range(start, stop, step)`

Третий параметр — шаг, на который меняется число:

- `range(2, 11, 3)` даёт `2, 5, 8`;
- следующее число `11` уже не входит.

![Визуализация range со стартом, остановкой и шагом](range_visual.png)

Проверим три формы вместе:

````python
print("range(4):")
for number in range(4):
    print(number)

print("range(2, 6):")
for number in range(2, 6):
    print(number)

print("range(2, 11, 3):")
for number in range(2, 11, 3):
    print(number)
````

### Считаем от 1 до N включительно

Поскольку `stop` не включается, для чисел от 1 до `N` пишут `range(1, N + 1)`. Например, если `N` равен 5, `range(1, 6)` выдаёт `1, 2, 3, 4, 5`.

Сначала значение из `input()` приходит как строка, поэтому используем уже знакомое преобразование `int()`.

````python
n = int(input("До какого числа считать? "))

for number in range(1, n + 1):
    print(number)
````

### Шаг может быть отрицательным

`range(5, 0, -1)` даёт 5, 4, 3, 2, 1. Правая граница 0 по-прежнему исключена. Знак шага должен вести от начала к границе: `range(1, 5, -1)` пустой. Нулевой шаг запрещён.

`range` не создаёт готовый список всех чисел. Он описывает диапазон, из которого цикл последовательно получает значения.


````python
for seconds in range(3, 0, -1):
    print(seconds)
print("Старт!")
for number in range(1, 4, -1):
    print("Эта строка не появится")
````

## 4. Следим за переменной: таблица значений

Полезный способ разобраться с циклом — составить таблицу. В каждой строке записывай очередное значение переменной и результат команды.

Для цикла ниже таблица такая:

| Итерация | `i` | `i * 2` |
|---:|---:|---:|
| 1 | 1 | 2 |
| 2 | 2 | 4 |
| 3 | 3 | 6 |
| 4 | 4 | 8 |

![Значение переменной цикла на каждой итерации](iteration_values.png)

Теперь проверь таблицу запуском:

````python
for i in range(1, 5):
    print(f"i = {i}; удвоенное значение = {i * 2}")
````

**Важное различие:** `i` меняет значение автоматически, потому что так работает цикл. Если внутри цикла написать `i = i + 1`, это не нужно для перехода к следующему числу и может запутать. Следующее значение выдаёт сам `range()`.

### Практика чтения

Заполни на бумаге таблицу для `range(3, 8, 2)`: запиши все значения, которые получит `n`. Затем проверь:

````python
for n in range(3, 8, 2):
    print(n)
````

## 5. От примеров к программам

### Программа 1: последовательность с шагом

Сначала пройдём вперёд с положительным шагом. Обратный отсчёт с отрицательным шагом разберём ниже.


````python
n = int(input("Введите N: "))

for number in range(1, n + 1):
    square = number * number
    print(f"{number} в квадрате = {square}")
````

Разбор: `n` хранит введённую границу; `range(1, n + 1)` выдаёт числа от 1 до `n`; переменная `number` указывает на текущую итерацию; `square` вычисляется заново для каждого числа.

### Программа 2: таблица умножения

Теперь пользователь вводит число, а цикл подставляет множители от 1 до 10.

````python
a = int(input("Какое число умножаем? "))

for multiplier in range(1, 11):
    result = a * multiplier
    print(f"{a} × {multiplier} = {result}")
````

### Программа 3: повторяем текст

`range()` может управлять количеством повторений, даже если переменная цикла не нужна для вывода. Здесь подчёркивание `_` — условное название, которое часто используют, когда значение шага не пригодится. Это всё ещё переменная цикла.

````python
name = input("Как тебя зовут? ")
count = int(input("Сколько раз поздороваться? "))

for _ in range(count):
    print(f"Привет, {name}!")
````

## 6. Подключаем знакомые условия `if / elif / else`

Условие можно поместить **внутрь** цикла. Тогда на каждой итерации проверяется очередное значение. Цикл выдаёт число, затем `if` решает, какое действие выполнить для этого числа.

![Схема сочетания for и if](for_if.png)

Пример: для чисел от 1 до 10 написать, чётное число или нечётное. Число чётное, если остаток от деления на 2 равен нулю: `number % 2 == 0`. Знак `%` и сравнения уже знакомы.

````python
for number in range(1, 11):
    if number % 2 == 0:
        print(number, "— чётное")
    else:
        print(number, "— нечётное")
````

Обрати внимание на отступы:

- тело цикла имеет один отступ;
- команды внутри `if` имеют два отступа;
- `else` выровнен с `if`, потому что это части одного условия;
- после ветки `if / else` цикл продолжает следующую итерацию.

Можно добавить `elif`, чтобы различать несколько случаев. Например, числа меньше 4 назвать маленькими, число 4 — особенным, остальные — большими.

````python
for number in range(1, 7):
    if number < 4:
        print(number, "— меньше четырёх")
    elif number == 4:
        print(number, "— ровно четыре")
    else:
        print(number, "— больше четырёх")
````

### Программа 4: делятся ли числа на 3

Пользователь задаёт верхнюю границу. Перебираем числа от 1 до неё и выводим только те, которые делятся на 3 без остатка. В этой программе цикл идёт по всем числам, а `if` выбирает, для каких чисел печатать сообщение.

````python
n = int(input("Введите верхнюю границу: "))

for number in range(1, n + 1):
    if number % 3 == 0:
        print(number, "делится на 3")
````

## 7. Частые ошибки и как себя проверить

1. **Забыли двоеточие:** после строки `for ...` нужен `:`.
2. **Не сделали отступ:** команды тела должны быть сдвинуты вправо.
3. **Ожидают включения `stop`:** `range(1, 5)` заканчивается на 4. Чтобы включить `N`, используй `range(1, N + 1)`.
4. **Начинают счёт не с того числа:** `range(5)` начинается с нуля; для счёта от 1 явно напиши `range(1, ...)`.
5. **Считают количество не так:** `range(5)` даёт пять чисел, хоть последнее и равно 4.
6. **Забыли `int()`:** результат `input()` — текст; для числовой границы нужно `int(input(...))`.
7. **Перепутали отступы `if` и `for`:** используй один уровень отступа для тела цикла и ещё один для тела условия.

Если результат неожиданный, выпиши значения `range()` по порядку и проследи тело для каждого значения.

## 8. Накопитель: результат растёт на каждом шаге

Нужно сложить числа от 1 до 4. Создай `total = 0` **перед** циклом, затем прибавляй текущее число. Запись `total += number` означает здесь то же, что `total = total + number`.

| Шаг | Число | Сумма до | Сумма после |
|---|---:|---:|---:|
| 1 | 1 | 0 | 1 |
| 2 | 2 | 1 | 3 |
| 3 | 3 | 3 | 6 |
| 4 | 4 | 6 | 10 |

Если поставить `total = 0` внутрь цикла, прошлый результат будет стираться. Если перенести итоговый `print()` внутрь, увидишь промежуточные суммы.


````python
total = 0
for number in range(1, 5):
    total += number
print("Сумма:", total)
````
