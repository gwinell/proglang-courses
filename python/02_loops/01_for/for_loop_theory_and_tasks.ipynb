## 1. Зачем нужны циклы

Цикл нужен, когда одну и ту же команду надо выполнить много раз. Вместо повторения одинаковых строк мы описываем действие один раз и говорим Python, сколько раз его повторить.

````python
print("Привет!")
print("Привет!")
print("Привет!")
````

## 2. Синтаксис `for`

Общий вид:
```python
for переменная in range(количество):
    команда
```

`for` начинает цикл, переменная получает очередное число, после `range(...)` ставится двоеточие, а команды внутри цикла имеют одинаковый отступ (обычно 4 пробела). Отступ показывает Python, какие команды относятся к циклу.

````python
for i in range(5):
    print("Повторение номер", i)
````

## 3. `range()` и правая граница

`range(N)` даёт числа от 0 до `N - 1`. Поэтому `range(5)` даёт 0, 1, 2, 3, 4 — всего пять чисел. Правая граница не включается.

`range(start, stop)` начинается с `start` и идёт до числа перед `stop`: `range(1, 6)` даёт 1, 2, 3, 4, 5.

`range(start, stop, step)` добавляет шаг: `range(2, 11, 2)` даёт 2, 4, 6, 8, 10. Для движения назад нужен отрицательный шаг: `range(10, 0, -2)`. 

````python
for number in range(1, 6):
    print(number)

print("Чётные:")
for number in range(2, 11, 2):
    print(number)
````

## 4. Пошаговое выполнение

Для `range(1, 4)` Python сначала берёт 1 и выполняет тело, затем 2, затем 3. После этого числа заканчиваются и цикл завершается. Число 4 не используется. Переменная цикла на каждом шаге получает новое значение.

````python
for number in range(1, 4):
    print("Начало, number =", number)
    print("Квадрат =", number * number)
    print("Конец")
````

## 5. Ввод до цикла и арифметика

Сначала можно получить число: `n = int(input(...))`, а затем использовать его в `range`. Для чисел от 1 до `n` обычно пишут `range(1, n + 1)`, потому что правая граница не включается. Внутри цикла разрешены обычные вычисления: сложение, вычитание, умножение, деление и остаток `%`.

````python
n = int(input("Введите N: "))
for number in range(1, n + 1):
    print(number, "-> квадрат", number * number)
````

## 6. `if / elif / else` внутри `for`

Условие внутри цикла проверяет каждое очередное число. Чётное число можно проверить так: `number % 2 == 0`. У `if` тоже есть двоеточие и отступ; команды внутри него получают ещё один отступ.

````python
for number in range(1, 11):
    if number % 2 == 0:
        print(number, "чётное")
    else:
        print(number, "нечётное")
````

## 7. Накопление суммы и подсчёт

Для суммы создай переменную до цикла: `summa = 0`, а внутри прибавляй: `summa = summa + number`. Для подсчёта подходящих чисел используй `count = 0` и увеличивай его на 1 внутри подходящего `if`. Не создавай накопитель внутри цикла — иначе он будет сбрасываться.

````python
n = int(input("Введите N: "))
summa = 0
count = 0
for number in range(1, n + 1):
    summa = summa + number
    if number % 2 == 0:
        count = count + 1
print("Сумма =", summa)
print("Чётных чисел =", count)
````

## Типичные ошибки

1. Забыто `:` после `for` или `if`.
2. Нет отступа у тела цикла.
3. Написано `range(1, n)`, хотя нужно включить `n`; используй `range(1, n + 1)`.
4. `input()` не превращён в число через `int()`.
5. Накопитель суммы создан внутри цикла.
6. Отрицательный шаг использован, когда начало меньше конца.
7. Перепутаны имена переменных.
