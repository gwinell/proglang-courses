## 1. Зачем нужен `while`

Цикл `for` удобен, когда заранее известно, сколько раз повторить действие или по каким значениям пройти. А если число повторений заранее неизвестно? Например, просить пароль до тех пор, пока он не станет правильным. Для этого подходит `while`.

`while` переводится как «пока». Он повторяет тело, **пока условие истинно (`True`)**.


````python
attempt = 1

while attempt <= 3:
    print("Попытка", attempt)
    attempt = attempt + 1

print("Попытки закончились")
````

В строке `while attempt <= 3:` записано условие. Команды с отступом составляют **тело цикла**. Полный проход тела называется **итерацией**. После каждой итерации Python снова проверяет условие.

## 2. Проверка условия происходит перед каждой итерацией

Порядок всегда такой:

1. Python проверяет условие.
2. Если оно истинно — выполняет тело цикла.
3. Доходит до конца тела и возвращается к проверке условия.
4. Если условие ложно — пропускает тело и идёт дальше после цикла.

![Схема проверки условия while](assets/while-check.png)

Условие проверяется **до** выполнения тела, а не после. Поэтому тело может не выполниться ни разу.


````python
number = 10

while number < 5:
    print("Эта строка не появится")

print("Программа продолжилась")
````

Здесь `10 < 5` сразу ложно. Python не заходит в тело цикла и сразу печатает сообщение после него. Это называется **ноль итераций**. Это нормальное поведение: число повторений зависит от условия.

````python
temperature = 3

while temperature > 0:
    print("Осталось:", temperature)
    temperature = temperature - 1

print("Ноль: стоп")
````

## 3. Как меняются переменные внутри цикла

Часто цикл `while` управляется переменной-счётчиком. Важно не только задать начальное значение, но и изменить его в теле так, чтобы условие однажды стало ложным.

В примере ниже `step` сначала равен `1`. После печати мы прибавляем к нему `1`. Следующая проверка использует уже новое значение.

![Пошаговая схема изменения счётчика](assets/counter-steps.png)


````python
step = 1

while step <= 4:
    print("Сейчас шаг", step)
    step = step + 1

print("Готово")
````

Проследим вручную:

| Проверка | `step` до проверки | `step <= 4` | Действие |
|---:|---:|---|---|
| 1 | 1 | истина | печать, затем `step` становится 2 |
| 2 | 2 | истина | печать, затем `step` становится 3 |
| 3 | 3 | истина | печать, затем `step` становится 4 |
| 4 | 4 | истина | печать, затем `step` становится 5 |
| 5 | 5 | ложь | тело пропускается, цикл завершён |

## 4. Бесконечный цикл и как его избежать

Бесконечный цикл — цикл, условие которого остаётся истинным всегда. Обычно это происходит, если забыть изменить переменную, от которой зависит условие, или изменить не в ту сторону.

Например, если `count` равен 1 и внутри цикла не меняется, условие `count <= 3` всегда будет истинным. Цикл будет бесконечно печатать строки.

![Визуализация бесконечного цикла и выхода](assets/infinite-loop.png)

Чтобы цикл завершился, проверь три вещи:

- начальное значение переменной;
- условие в строке `while`;
- изменение переменной внутри тела: оно должно приближать условие к ложному.


````python
count = 1

while count <= 3:
    print("Повтор", count)
    count = count + 1  # изменение не даёт циклу повторяться бесконечно

print("Цикл завершён")
````

## 5. `while` и ввод пользователя

Теперь условие цикла зависит от ответа человека. Пока пользователь вводит не то слово, программа просит повторить. После правильного ответа условие становится ложным.


````python
answer = input("Напиши слово старт: ")

while answer != "старт":
    print("Пока не получилось. Попробуй ещё раз.")
    answer = input("Напиши слово старт: ")

print("Отлично, начинаем!")
````

Обрати внимание: `input()` находится **внутри** цикла. Поэтому на каждой итерации можно получить новый ответ. Если оставить ввод только снаружи, ответ не будет меняться, и при неправильном ответе цикл не остановится.

## 6. Модуль `random`: случайные числа

Модуль — это готовый набор возможностей Python. Модуль `random` умеет выбирать случайные значения. Чтобы воспользоваться им, сначала подключим модуль:

```python
import random
```

После этого функция вызывается с именем модуля: `random.randint(1, 6)`. `randint(a, b)` возвращает случайное целое число от `a` до `b` **включительно**: обе границы могут выпасть.

![Схема randint как выбора числа из диапазона](assets/randint-range.png)


````python
import random

number = random.randint(1, 6)
print("Выпало число", number)
````

Запусти ячейку несколько раз. Результат может быть разным; случайность не означает, что каждое число обязательно встретится за шесть запусков.

### Проверь границы

## 7. `choice()`: случайно выбрать символ из строки

`random.choice(строка)` выбирает один случайный символ из строки. Строка — это знакомый нам текст, например `"КОТ"`; функция может вернуть `"К"`, `"О"` или `"Т"`. Сейчас используем `choice()` только со строками и не вводим списки.

![Схема choice как случайного выбора одного варианта](assets/choice-option.png)


````python
letter = random.choice("ABC")

if letter == "A":
    drink = "чай"
elif letter == "B":
    drink = "какао"
else:
    drink = "сок"

print("Случайно выбрана буква", letter)
print("Сегодня выбираем:", drink)
````

Каждый запуск функция выбирает один символ строки `"ABC"`. Переменная `letter` получает строку из одного символа. Затем знакомые `if / elif / else` превращают букву в название напитка. Так мы выбираем один из трёх вариантов, работая только со строкой.

## 8. Соединяем `while` со случайностью и условиями

Построим игру «Угадай число». Программа выбирает число, а игрок пробует угадать. Цикл продолжается, пока ответ не совпал с загаданным числом. Внутри цикла `if / elif / else` сообщает, больше или меньше догадка.

![Схема программы while вместе с if](assets/while-if.png)


````python
import random

secret = random.randint(1, 10)
guess = int(input("Угадай число от 1 до 10: "))

while guess != secret:
    if guess < secret:
        print("Моё число больше")
    else:
        print("Моё число меньше")

    guess = int(input("Попробуй ещё раз: "))

print("Верно! Ты угадал.")
````

Здесь `guess != secret` — условие продолжения игры. Если догадка не равна загаданному числу, тело выполняется. После подсказки программа снова спрашивает число — переменная `guess` меняется. Когда значения совпадут, условие станет ложным и игра закончится.

## 9. Программа с несколькими условиями и логикой

Сделаем меню: программа случайно выбирает задание, а пользователь может попросить новое. Цикл работает, пока ответ равен `да`. Внутренние условия проверяют ответ.


````python
again = "да"

while again == "да":
    task_letter = random.choice("ABC")
    if task_letter == "A":
        task = "назови число больше 10"
    elif task_letter == "B":
        task = "посчитай 7 + 8"
    else:
        task = "назови чётное число"
    print("Задание:", task)
    again = input("Ещё одно задание? Напиши да или нет: ")

print("На сегодня всё!")
````

Можно сделать условие цикла терпимее к двум ответам: продолжать, если пользователь написал `да` **или** `ещё`. Это знакомый оператор `or`.


````python
again = input("Продолжить? ")

while again == "да" or again == "ещё":
    print("Продолжаем!")
    again = input("Напиши да, ещё или нет: ")

print("Завершили")
````

`and`, `or` и `not` работают в условии `while` так же, как в `if`. Например, условие `age >= 10 and age <= 15` истинно, когда возраст входит в диапазон от 10 до 15.

## 10. Досрочная остановка и пропуск шага

`break` сразу завершает ближайший цикл. `continue` пропускает оставшиеся команды текущей итерации и возвращает выполнение к следующей проверке условия.

Сначала проследи, где меняется счётчик. В примере он увеличивается **до** `continue`, поэтому пропуск не зацикливает программу.


````python
number = 0
while number < 6:
    number += 1
    if number == 2:
        continue
    if number == 5:
        break
    print(number)
print("После цикла:", number)
````

**Предскажи.** Удали условие с `break`. Какими станут вывод и последнее значение `number`? Затем проверь. В цикле `for` эти две команды работают так же.

**Дополнительно: `else` у цикла.** Он выполняется, если цикл закончился без `break`: у `while` условие стало ложным, у `for` закончились элементы. Это относится и к циклу с нулём итераций. Команды после всего цикла выполняются и после `break`.

Проверь следующий пример для `target = 3`, затем для `target = 8`.


````python
target = 3
for number in range(1, 5):
    if number == target:
        print("Нашли:", number)
        break
else:
    print("Не нашли")
print("Поиск завершён")
````
