# Вложенные циклы `for`

Когда для каждого элемента нужно выполнить ещё несколько повторов, один цикл можно поместить внутрь другого. Внешний цикл выбирает строку, внутренний выполняет действия в этой строке.

**Предпосылки:** `for`, `range()`, вывод `print()`, отступы. Здесь мы работаем только со счётчиками и числами.

## Результаты урока

Ты сможешь проследить порядок внешнего и внутреннего цикла; сосчитать, сколько раз выполнится внутреннее тело; построить прямоугольник и растущую лестницу; объяснить, где переносится строка.


## 1. Вложенность и порядок

Посмотри на значения `row` и `column`. Для каждой строки внутренний цикл проходит все столбцы от начала до конца. Только после этого внешний цикл выбирает следующую строку.


```python
for row in range(2):
    for column in range(3):
        print("строка", row, "столбец", column)
```

До запуска выпиши все пары `(row, column)` в том порядке, в каком их покажет Python. Затем запусти ячейку и сверь список. Внешний цикл выполнится 2 раза, внутреннее тело — `2 * 3 = 6` раз.

Если внутренний диапазон содержит 0 значений, его тело не выполнится ни разу для каждой строки. Если внешний диапазон пустой, внутренний цикл не начнётся вообще.


## 2. Прямоугольник

Нужно напечатать 3 строки, в каждой по 5 звёздочек. Внешний цикл отвечает за количество строк. Внутренний цикл собирает символы одной строки. Здесь `end=""` просит `print()` не переходить на следующую строку после каждого символа. Последний `print()` завершает строку.


```python
height = 3
width = 5

for row in range(height):
    for column in range(width):
        print("*", end="")
    print()
```

**Предскажи:** сколько раз выполнится строка `print("*", end="")`? Что изменится, если последний `print()` сдвинуть на один отступ вправо? Проверь обе версии и объясни.

**Измени:** установи `height = 1`, затем `width = 1`. Размеры должны совпадать с числом строк и символов. При нулевой ширине внутреннее тело ничего не печатает, но внешний цикл всё ещё вызывает `print()` — появятся пустые строки.

## 3. Лестница

В растущей лестнице каждая строка длиннее предыдущей на один символ. Внешний цикл задаёт номер строки. Внутренний печатает столько символов, сколько соответствует этой строке.


```python
height = 4

for row in range(1, height + 1):
    for column in range(row):
        print("#", end="")
    print()
```

Измени условие диапазона внутреннего цикла на `range(height)`. Предскажи, получится лестница или прямоугольник. Запусти и объясни, почему каждая строка теперь одинаковой длины.

## 4. Таблица умножения

Внешний цикл выбирает первый множитель, внутренний — второй. Параметры `start` и `stop` у `range()` следуют уже изученному правилу: правая граница не включается.


```python
for first in range(1, 4):
    for second in range(1, 4):
        print(first * second, end=" ")
    print()
```

**Задание на чтение.** Сколько значений напечатает программа? Для каждого результата назови пару множителей. Затем измени внутренний цикл так, чтобы вывести только пары, в которых второй множитель не меньше первого. Используй знакомый `if`; не создавай пока список пар.

## Частые ошибки

- Внутренний цикл случайно расположен на одном уровне с внешним из-за отступа.
- Внутренний цикл использует предел `height`, хотя для прямоугольника нужен `width`.
- `print()` для перехода на новую строку расположен внутри внутреннего цикла.
- Граница `range()` включена ошибочно; чтобы включить `height`, обычно нужен `height + 1`.
- Считают, что общее число проходов — только число внешних повторов. Для прямоугольной таблицы это произведение размеров.

## Критерий готовности

## Связь с дальнейшими темами
